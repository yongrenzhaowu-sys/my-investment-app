```python
# ===================================
# リターン計算関数の修正版（投資比率を考慮した正しい計算）
# ===================================

def calculate_long_portfolio_return_with_units_fixed(
    portfolio: dict,
    start_date: pd.Timestamp, 
    end_date: pd.Timestamp, 
    prices_df: pd.DataFrame,
    initial_capital: float
) -> dict:
    """100株単位ポートフォリオのリターン計算（修正版）"""
    
    if not portfolio['stocks']:
        return {
            'gross_return': 0.0,
            'net_return': 0.0,
            'tax': 0.0,
            'total_investment': 0.0,
            'cash_remaining': initial_capital,
            'investment_ratio': 0.0
        }
    
    stocks = portfolio['stocks']
    start_shares = portfolio['shares']
    start_prices = portfolio['prices']
    start_amounts = portfolio['amounts']
    
    total_investment = sum(start_amounts)
    cash_remaining = initial_capital - total_investment
    
    # 終了時点の価格取得
    end_window = prices_df[
        (prices_df['Code'].isin(stocks)) &
        (prices_df['Date'] >= end_date - pd.Timedelta(days=5)) &
        (prices_df['Date'] <= end_date + pd.Timedelta(days=5))
    ].sort_values(['Code', 'Date']).groupby('Code').last()
    
    # 各銘柄の損益計算
    total_profit = 0.0
    total_start_value = 0.0
    
    for i, code in enumerate(stocks):
        if code not in end_window.index:
            continue
        
        shares = start_shares[i]
        start_price = start_prices[i]
        end_price = end_window.loc[code, 'Close']
        
        start_value = shares * start_price
        end_value = shares * end_price
        profit = end_value - start_value
        
        total_start_value += start_value
        total_profit += profit
    
    if total_start_value == 0:
        return {
            'gross_return': 0.0,
            'net_return': 0.0,
            'tax': 0.0,
            'total_investment': total_investment,
            'cash_remaining': cash_remaining,
            'investment_ratio': 0.0
        }
    
    # 【修正】投資額に対するリターン（ポートフォリオリターン）を計算
    # これが正しいリターン計算方法です
    if total_start_value > 0:
        portfolio_return = total_profit / total_start_value
    else:
        portfolio_return = 0.0
    
    # 税金計算（利益のみ課税）
    taxable_profit = max(total_profit, 0)
    tax = taxable_profit * TAX_RATE
    
    # 税引後利益
    net_profit = total_profit - tax
    
    # 投資額に対する税引後リターン
    if total_start_value > 0:
        portfolio_net_return = net_profit / total_start_value
    else:
        portfolio_net_return = 0.0
    
    # 【修正】資本全体に対するリターン（投資比率を考慮）
    # 投資額に対するリターン × 投資比率 = 資本全体に対するリターン
    # これにより、複利計算が正しく行われます
    investment_ratio = total_start_value / initial_capital if initial_capital > 0 else 0.0
    gross_return_total = portfolio_return * investment_ratio
    net_return_total = portfolio_net_return * investment_ratio
    tax_rate_total = (tax / total_start_value) * investment_ratio if total_start_value > 0 else 0.0
    
    return {
        'gross_return': gross_return_total,
        'net_return': net_return_total,
        'tax': tax_rate_total,
        'total_investment': total_investment,
        'cash_remaining': cash_remaining,
        'investment_ratio': total_investment / initial_capital
    }

# 修正版の関数を元の関数名に置き換え
calculate_long_portfolio_return_with_units = calculate_long_portfolio_return_with_units_fixed

```

```python
import pandas as pd
import numpy as np
from tqdm import tqdm 
import warnings
import os
import logging
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

warnings.filterwarnings('ignore')

# ===================================
# ロギング設定
# ===================================
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[
        logging.FileHandler('backtest_long_only_october_unit_constraints.log', encoding='utf-8'),
        logging.StreamHandler()
    ]
)
logger = logging.getLogger(__name__)

# ===================================
# グローバル変数
# ===================================
CACHE_FILE = 'topix_quarterly_statements.csv'
PRICE_CACHE = {}

# 税金パラメータ
TAX_RATE = 0.20315  # 譲渡益税 20.315%

# 単位株制限
UNIT_SHARES = 100  # 100株単位

# ===================================
# A. 財務データ読み込み（列名自動判定版）
# ===================================

def load_financial_data(cache_filename: str = CACHE_FILE) -> pd.DataFrame:
    """財務諸表データをキャッシュから読み込み（列名を自動判定）"""
    
    if not os.path.exists(cache_filename):
        logger.error(f"財務データファイル '{cache_filename}' が見つかりません")
        return pd.DataFrame()
    
    try:
        df = pd.read_csv(
            cache_filename, 
            encoding='utf-8-sig',
            parse_dates=['DisclosedDate'],
            low_memory=False
        )
        
        logger.info(f"財務データ読み込み成功: {len(df):,}件")
        
        # 列名のマッピング
        column_mapping = {
            'IssuedShareTotal': ['IssuedShareTotal', 'NumberOfIssuedAndOutstandingSharesAtTheEndOfFiscalYearIncludingTreasuryStock'],
            'Equity': ['Equity', 'NetAssets', 'TotalEquity'],
            'Profit': ['Profit', 'NetIncome', 'ProfitAttributableToOwnersOfParent']
        }
        
        available_cols = df.columns.tolist()
        required_columns = {}
        
        for target_col, possible_names in column_mapping.items():
            found = False
            for possible_name in possible_names:
                if possible_name in available_cols:
                    required_columns[target_col] = possible_name
                    found = True
                    break
            if not found:
                required_columns[target_col] = None
        
        rename_dict = {v: k for k, v in required_columns.items() if v is not None}
        df = df.rename(columns=rename_dict)
        
        for col in ['IssuedShareTotal', 'Equity', 'Profit']:
            if col not in df.columns:
                df[col] = 0
        
        base_cols = ['Code', 'DisclosedDate']
        if 'CompanyName' in df.columns:
            base_cols.append('CompanyName')
        
        final_cols = base_cols + ['Profit', 'Equity', 'IssuedShareTotal']
        df = df[final_cols].copy()
        
        logger.info(f"使用列: {df.columns.tolist()}")
        
        return df
    
    except Exception as e:
        logger.error(f"財務データ読み込みエラー: {e}", exc_info=True)
        return pd.DataFrame()

# ===================================
# B. 株価データ読み込み（メモリ効率化版）
# ===================================

def load_existing_price_data(ohlcv_dir: str = './OHLCV_Adjusted') -> pd.DataFrame:
    """株価データ読み込み（メモリ効率化版）"""
    global PRICE_CACHE
    
    if not os.path.exists(ohlcv_dir):
        logger.error(f"株価データディレクトリ '{ohlcv_dir}' が見つかりません。")
        return pd.DataFrame()
    
    csv_files = sorted([
        f for f in os.listdir(ohlcv_dir) 
        if f.startswith('OHLCV_Adjusted_') and f.endswith('.csv') and f != 'OHLCV_Adjusted_TOPIX.csv'
    ])
    
    if not csv_files:
        logger.error(f"ディレクトリ '{ohlcv_dir}' 内にCSVファイルが見つかりません。")
        return pd.DataFrame()
    
    logger.info(f"株価ファイル数: {len(csv_files)}個")
    
    all_dataframes = []
    usecols = ['Date', 'Ticker', 'AdjustmentClose']
    
    for csv_file in tqdm(csv_files, desc="株価ファイル読み込み中"):
        file_path = os.path.join(ohlcv_dir, csv_file)
        try:
            df = pd.read_csv(
                file_path,
                usecols=usecols,
                parse_dates=['Date'],
                dtype={'Ticker': 'Int64', 'AdjustmentClose': 'float32'}
            )
            
            df = df.drop_duplicates(subset=['Ticker', 'Date'], keep='first')
            all_dataframes.append(df)
            
        except Exception as e:
            logger.warning(f"ファイル読み込みエラー ({csv_file}): {e}")
            continue
    
    if not all_dataframes:
        return pd.DataFrame()
    
    logger.info("全ファイルを結合中...")
    df_all = pd.concat(all_dataframes, ignore_index=True)
    
    logger.info(f"結合完了: {len(df_all):,}件")
    
    df_all = df_all.rename(columns={'Ticker': 'Code', 'AdjustmentClose': 'Close'})
    df_all['Code'] = df_all['Code'].astype('str').str.replace('<NA>', '0').str.zfill(4)
    df_all = df_all.dropna(subset=['Close'])
    
    logger.info("重複除去 & ソート中...")
    df_all = df_all.sort_values(['Code', 'Date'])
    df_all = df_all.drop_duplicates(subset=['Code', 'Date'], keep='first')
    
    logger.info(f"重複除去後: {len(df_all):,}件")
    
    return df_all

# ===================================
# C. 財務指標計算（チャンク処理版）
# ===================================

def safe_code_to_int(code_series: pd.Series) -> pd.Series:
    """証券コードを整数に変換（ゼロ埋め対応）"""
    cleaned = code_series.astype(str).str.replace(r'\D', '', regex=True)
    cleaned = cleaned.replace('', '0')
    return pd.to_numeric(cleaned, errors='coerce').fillna(0).astype('int64')

def calculate_market_metrics_fast_chunked(statements_df: pd.DataFrame, 
                                          prices_df: pd.DataFrame,
                                          chunk_size: int = 200) -> pd.DataFrame:
    """時価総額・PBR・ROE計算（チャンク処理版）"""
    
    logger.info("時価総額・PBR・ROE計算中（チャンク処理版）...")
    
    # prices_dfが空または必要な列が存在しない場合のチェック
    if prices_df.empty:
        logger.error("株価データが空です。株価データを読み込んでください。")
        return pd.DataFrame()
    
    if 'Code' not in prices_df.columns or 'Date' not in prices_df.columns or 'Close' not in prices_df.columns:
        logger.error(f"株価データに必要な列が存在しません。存在する列: {prices_df.columns.tolist()}")
        return pd.DataFrame()
    
    statements_df = statements_df.copy()
    statements_df['Profit'] = pd.to_numeric(statements_df['Profit'], errors='coerce').fillna(0)
    statements_df['Equity'] = pd.to_numeric(statements_df['Equity'], errors='coerce').fillna(0)
    statements_df['IssuedShareTotal'] = pd.to_numeric(statements_df['IssuedShareTotal'], errors='coerce').fillna(1)
    
    statements_df = statements_df[
        (statements_df['Equity'] > 0) & 
        (statements_df['IssuedShareTotal'] > 0)
    ]
    logger.info(f"有効な財務データ: {len(statements_df):,}件")
    
    # Code変換
    logger.info("Code列を整数型に変換中...")
    statements_df['Code_int'] = safe_code_to_int(statements_df['Code'])
    prices_df['Code_int'] = safe_code_to_int(prices_df['Code'])
    
    statements_df = statements_df[statements_df['Code_int'] > 0]
    prices_df = prices_df[prices_df['Code_int'] > 0]
    
    # ソート
    statements_df = statements_df.sort_values(['Code_int', 'DisclosedDate']).reset_index(drop=True)
    prices_df = prices_df.sort_values(['Code_int', 'Date']).reset_index(drop=True)
    
    # 重複除去
    statements_df = statements_df.drop_duplicates(subset=['Code_int', 'DisclosedDate'], keep='first')
    prices_df = prices_df.drop_duplicates(subset=['Code_int', 'Date'], keep='first')
    
    logger.info(f"ソート・重複除去後: 財務 {len(statements_df):,}件, 株価 {len(prices_df):,}件")
    
    # 銘柄ごとにマージ
    logger.info("銘柄ごとにマージ中...")
    
    statements_groups = list(statements_df.groupby('Code_int'))
    prices_dict = {code: group for code, group in prices_df.groupby('Code_int')}
    
    merged_list = []
    num_chunks = (len(statements_groups) + chunk_size - 1) // chunk_size
    
    for chunk_idx in tqdm(range(num_chunks), desc="マージ処理"):
        start_idx = chunk_idx * chunk_size
        end_idx = min((chunk_idx + 1) * chunk_size, len(statements_groups))
        
        chunk_groups = statements_groups[start_idx:end_idx]
        
        for code, stmt_code in chunk_groups:
            if code not in prices_dict:
                continue
            
            price_code = prices_dict[code]
            
            if len(price_code) == 0:
                continue
            
            stmt_code = stmt_code.sort_values('DisclosedDate').reset_index(drop=True)
            price_code = price_code.sort_values('Date').reset_index(drop=True)
            
            try:
                merged = pd.merge_asof(
                    stmt_code,
                    price_code[['Date', 'Close']],
                    left_on='DisclosedDate',
                    right_on='Date',
                    direction='backward',  # 【修正】ルックアヘッドバイアス回避
                    tolerance=pd.Timedelta(days=10)
                )
                
                if not merged.empty:
                    merged_list.append(merged)
            except Exception:
                continue
    
    if not merged_list:
        logger.error("マージ結果が空です")
        return pd.DataFrame()
    
    df_merged = pd.concat(merged_list, ignore_index=True)
    df_merged = df_merged.dropna(subset=['Close'])
    
    logger.info(f"マージ完了: {len(df_merged):,}件")
    
    # 財務指標計算
    df_merged['MarketCap'] = df_merged['Close'] * df_merged['IssuedShareTotal']
    df_merged['PBR'] = df_merged['MarketCap'] / df_merged['Equity']
    df_merged['ROE'] = (df_merged['Profit'] / df_merged['Equity']) * 100
    
    result_cols = ['Code', 'DisclosedDate', 'Close', 'MarketCap', 'PBR', 'ROE', 'Date']
    
    if 'CompanyName' in df_merged.columns:
        result_cols.insert(1, 'CompanyName')
    
    result_df = df_merged[result_cols].copy()
    result_df = result_df.rename(columns={'Close': 'StockPrice', 'Date': 'PriceDate'})
    
    # 異常値除去
    mask = (
        (result_df['PBR'] > 0) &
        (result_df['PBR'] < 50) &
        (result_df['ROE'] > -100) &
        (result_df['ROE'] < 100) &
        (result_df['MarketCap'] > 1_000_000_000)
    )
    
    result_df = result_df[mask].copy()
    
    logger.info(f"計算完了: {len(result_df):,}件")
    
    return result_df

# ===================================
# D. ポートフォリオ構築（100株単位制限版）
# ===================================

def build_unit_share_portfolio(stock_candidates: pd.DataFrame,
                               target_positions: int = 20,
                               initial_capital: float = 10_000_000) -> dict:
    """
    100株単位ポートフォリオ構築（実運用制約版）
    
    Args:
        stock_candidates: 候補銘柄データ
        target_positions: 目標銘柄数
        initial_capital: 初期資金
    
    Returns:
        dict: {
            'stocks': 選択された銘柄コードのリスト,
            'shares': 各銘柄の株数,
            'prices': 各銘柄の価格,
            'amounts': 各銘柄の投資額
        }
    """
    
    if len(stock_candidates) == 0:
        return {
            'stocks': [],
            'shares': [],
            'prices': [],
            'amounts': []
        }
    
    # 上位N銘柄を選択
    selected = stock_candidates.head(target_positions).copy()
    
    # 1銘柄あたりの配分資金
    capital_per_stock = initial_capital / len(selected)
    
    stocks = []
    shares_list = []
    prices_list = []
    amounts_list = []
    
    for _, row in selected.iterrows():
        code = row['Code']
        price = row['StockPrice']
        
        # 100株単位で購入可能な株数を計算
        required_amount = price * UNIT_SHARES
        
        if required_amount <= capital_per_stock:
            shares = int(capital_per_stock // required_amount) * UNIT_SHARES
            
            if shares > 0:
                stocks.append(code)
                shares_list.append(shares)
                prices_list.append(price)
                amounts_list.append(shares * price)
    
    return {
        'stocks': stocks,
        'shares': shares_list,
        'prices': prices_list,
        'amounts': amounts_list
    }

# ===================================
# E. リターン計算（100株単位版）
# ===================================

def calculate_long_portfolio_return_with_units(
    portfolio: dict,
    start_date: pd.Timestamp, 
    end_date: pd.Timestamp, 
    prices_df: pd.DataFrame,
    initial_capital: float
) -> dict:
    """100株単位ポートフォリオのリターン計算"""
    
    if not portfolio['stocks']:
        return {
            'gross_return': 0.0,
            'net_return': 0.0,
            'tax': 0.0,
            'total_investment': 0.0,
            'cash_remaining': initial_capital,
            'investment_ratio': 0.0
        }
    
    stocks = portfolio['stocks']
    start_shares = portfolio['shares']
    start_prices = portfolio['prices']
    start_amounts = portfolio['amounts']
    
    total_investment = sum(start_amounts)
    cash_remaining = initial_capital - total_investment
    
    # 終了時点の価格取得
    end_window = prices_df[
        (prices_df['Code'].isin(stocks)) &
        (prices_df['Date'] >= end_date - pd.Timedelta(days=5)) &
        (prices_df['Date'] <= end_date + pd.Timedelta(days=5))
    ].sort_values(['Code', 'Date']).groupby('Code').last()
    
    # 各銘柄の損益計算
    total_profit = 0.0
    total_start_value = 0.0
    
    for i, code in enumerate(stocks):
        if code not in end_window.index:
            continue
        
        shares = start_shares[i]
        start_price = start_prices[i]
        end_price = end_window.loc[code, 'Close']
        
        start_value = shares * start_price
        end_value = shares * end_price
        profit = end_value - start_value
        
        total_start_value += start_value
        total_profit += profit
    
    if total_start_value == 0:
        return {
            'gross_return': 0.0,
            'net_return': 0.0,
            'tax': 0.0,
            'total_investment': total_investment,
            'cash_remaining': cash_remaining,
            'investment_ratio': 0.0
        }
    
    # 税引前リターン（全資本ベース）
    gross_return_total = total_profit / initial_capital
    
    # 税金計算（利益のみ課税）
    taxable_profit = max(total_profit, 0)
    tax = taxable_profit * TAX_RATE
    
    # 税引後リターン（全資本ベース）
    net_profit = total_profit - tax
    net_return_total = net_profit / initial_capital
    tax_rate_total = tax / initial_capital
    
    return {
        'gross_return': gross_return_total,
        'net_return': net_return_total,
        'tax': tax_rate_total,
        'total_investment': total_investment,
        'cash_remaining': cash_remaining,
        'investment_ratio': total_investment / initial_capital
    }

def calculate_topix_return(start_date: pd.Timestamp, end_date: pd.Timestamp, 
                          prices_df: pd.DataFrame) -> float:
    """TOPIX基準リターン計算（全銘柄等ウェイト）"""
    all_stocks = prices_df['Code'].unique()[:100].tolist()
    
    # 開始時点の価格
    start_window = prices_df[
        (prices_df['Code'].isin(all_stocks)) &
        (prices_df['Date'] >= start_date - pd.Timedelta(days=5)) &
        (prices_df['Date'] <= start_date + pd.Timedelta(days=5))
    ].sort_values(['Code', 'Date']).groupby('Code').first()
    
    # 終了時点の価格
    end_window = prices_df[
        (prices_df['Code'].isin(all_stocks)) &
        (prices_df['Date'] >= end_date - pd.Timedelta(days=5)) &
        (prices_df['Date'] <= end_date + pd.Timedelta(days=5))
    ].sort_values(['Code', 'Date']).groupby('Code').last()
    
    common_codes = start_window.index.intersection(end_window.index)
    
    if len(common_codes) == 0:
        return 0.0
    
    start_prices = start_window.loc[common_codes, 'Close'].values
    end_prices = end_window.loc[common_codes, 'Close'].values
    returns = (end_prices - start_prices) / start_prices
    
    return np.mean(returns)

# ===================================
# F. バックテスト（10月1日リバランス・100株単位版）
# ===================================

def smbc_value_quality_backtest_october_unit(enhanced_financial_data: pd.DataFrame, 
                                            prices_df: pd.DataFrame,
                                            initial_capital: float = 10_000_000) -> pd.DataFrame:
    """SMBC割安高質戦略（10月1日リバランス・100株単位制限版）"""
    
    logger.info(f"10月1日リバランス戦略バックテスト実行中（100株単位・初期資本{initial_capital:,.0f}円）...")
    
    # 10月1日のリバランス日を生成（2016年10月〜2025年10月）
    rebalance_dates = [pd.Timestamp(f'{year}-10-01') for year in range(2016, 2026)]
    
    logger.info(f"リバランス日数: {len(rebalance_dates)}回")
    logger.info(f"リバランス日: {[d.strftime('%Y-%m-%d') for d in rebalance_dates]}")
    
    strategy_results = []
    
    for i, rebalance_date in enumerate(tqdm(rebalance_dates[:-1], desc="10月1日リバランス戦略（100株単位）")):
        
        next_rebalance = rebalance_dates[i+1]
        
        current_data = enhanced_financial_data[
            enhanced_financial_data['DisclosedDate'] <= rebalance_date
        ].copy()
        
        current_data = current_data.sort_values('DisclosedDate').groupby('Code').tail(1)
        
        if len(current_data) < 100:
            logger.warning(f"{rebalance_date}: データ不足（{len(current_data)}銘柄）")
            continue
        
        # PBRとROEでランキング
        current_data['PBR_Rank'] = current_data['PBR'].rank(method='first', ascending=True)
        current_data['ROE_Rank'] = current_data['ROE'].rank(method='first', ascending=False)
        
        # 四分位分割
        current_data['PBR_Quartile'] = pd.qcut(current_data['PBR_Rank'], q=4, labels=[1, 2, 3, 4])
        current_data['ROE_Quartile'] = pd.qcut(current_data['ROE_Rank'], q=4, labels=[1, 2, 3, 4])
        
        # ロング候補：低PBR × 高ROE
        long_candidates = current_data[
            (current_data['PBR_Quartile'] == 1) & 
            (current_data['ROE_Quartile'] == 4)
        ].nsmallest(50, 'PBR')  # 上位50銘柄から選択
        
        # 100株単位ポートフォリオ構築
        long_portfolio = build_unit_share_portfolio(
            long_candidates,
            target_positions=20,
            initial_capital=initial_capital
        )
        
        logger.info(f"{rebalance_date.strftime('%Y-%m')}: {len(long_portfolio['stocks'])}銘柄選定、"
                   f"投資額{sum(long_portfolio['amounts']):,.0f}円")
        
        # リターン計算
        long_result = calculate_long_portfolio_return_with_units(
            long_portfolio, rebalance_date, next_rebalance, prices_df, initial_capital
        )
        
        topix_return = calculate_topix_return(rebalance_date, next_rebalance, prices_df)
        
        strategy_results.append({
            'date': next_rebalance,
            'strategy_return_gross': long_result['gross_return'],
            'strategy_return_net': long_result['net_return'],
            'long_return_gross': long_result['gross_return'],
            'long_return_net': long_result['net_return'],
            'tax': long_result['tax'],
            'topix_return': topix_return,
            'long_count': len(long_portfolio['stocks']),
            'investment_ratio': long_result['investment_ratio']
        })
    
    return pd.DataFrame(strategy_results)

# ===================================
# G. パフォーマンス分析
# ===================================

def calculate_annual_performance(returns: pd.Series, dates: pd.Series = None) -> float:
    """
    年率リターン計算
    
    Args:
        returns: 期間ごとのリターン
        dates: 期間の日付（オプション、日次リバランスの場合に使用）
    """
    if len(returns) == 0:
        return 0.0
    
    cumulative = (1 + returns).prod()
    
    # 日付が提供されている場合は、実際の年数を使用
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
    else:
        # 日付が提供されていない場合は、期間数を年数として扱う（年次リバランスの場合）
        years = len(returns)
    
    if years > 0:
        return ((cumulative ** (1/years)) - 1) * 100
    else:
        return 0.0

def calculate_volatility_annualized(returns: pd.Series, dates: pd.Series = None) -> float:
    """年率換算ボラティリティ計算"""
    if len(returns) == 0:
        return 0.0
    
    period_std = returns.std()
    
    # 日付が提供されている場合は、実際の年数を使用
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
        periods_per_year = len(returns) / years if years > 0 else len(returns)
    else:
        # 日付が提供されていない場合は、期間数を年数として扱う（年次リバランスの場合）
        periods_per_year = 1
    
    if periods_per_year > 0:
        return period_std * np.sqrt(periods_per_year) * 100
    else:
        return 0.0

def calculate_sharpe_ratio(returns: pd.Series, risk_free_rate: float = 0.0, dates: pd.Series = None) -> float:
    """シャープレシオ計算（年率換算）"""
    if len(returns) == 0 or returns.std() == 0:
        return 0.0
    
    excess_returns = returns - risk_free_rate
    mean_return = excess_returns.mean()
    std_return = excess_returns.std()
    
    # 日付が提供されている場合は、実際の年数を使用
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
        periods_per_year = len(returns) / years if years > 0 else len(returns)
    else:
        # 日付が提供されていない場合は、期間数を年数として扱う（年次リバランスの場合）
        periods_per_year = 1
    
    if std_return > 0:
        return (mean_return / std_return) * np.sqrt(periods_per_year)
    else:
        return 0.0

def calculate_sortino_ratio(returns: pd.Series, target: float = 0.0) -> float:
    """ソルティノレシオ計算"""
    excess_returns = returns - target
    downside_returns = excess_returns[excess_returns < 0]
    
    # 損失がない場合や損失が1つしかない場合は適切に処理
    if len(downside_returns) == 0:
        # 損失がない場合は無限大を返す（または適切な値）
        return np.inf if excess_returns.mean() > 0 else 0.0
    elif len(downside_returns) == 1:
        # 損失が1つしかない場合、標準偏差の代わりに絶対値を使用
        downside_std = abs(downside_returns.iloc[0])
    else:
        downside_std = downside_returns.std()
    
    if downside_std > 0:
        return (excess_returns.mean() / downside_std) * np.sqrt(1)
    else:
        return 0.0

def calculate_calmar_ratio(returns: pd.Series, dates: pd.Series = None) -> float:
    """カルマーレシオ計算"""
    annual_return = calculate_annual_performance(returns, dates)
    mdd_result = calculate_max_drawdown(returns, dates)
    max_dd = abs(mdd_result['max_drawdown'])
    return annual_return / max_dd if max_dd > 0 else 0

def calculate_max_drawdown(returns: pd.Series, dates: pd.Series = None) -> dict:
    """
    最大ドローダウン計算（発生日も返す）
    
    Args:
        returns: 期間ごとのリターン
        dates: 期間の日付（オプション）
    
    Returns:
        dict: {
            'max_drawdown': 最大ドローダウン（%）,
            'mdd_date': MDD発生日（pd.TimestampまたはNone）,
            'mdd_period': MDD発生期間の文字列
        }
    """
    if len(returns) == 0:
        return {
            'max_drawdown': 0.0,
            'mdd_date': None,
            'mdd_period': None
        }
    
    cumulative = (1 + returns).cumprod()
    running_max = cumulative.cummax()
    drawdown = (cumulative - running_max) / running_max
    
    min_dd_idx = drawdown.idxmin()
    max_dd_value = drawdown.min() * 100
    
    # 発生日を特定
    mdd_date = None
    mdd_period = None
    
    if dates is not None and len(dates) > 0:
        # インデックスが一致するか、位置（整数）として扱う
        if min_dd_idx in dates.index:
            mdd_date = dates.loc[min_dd_idx]
        elif isinstance(min_dd_idx, (int, np.integer)) and 0 <= min_dd_idx < len(dates):
            mdd_date = dates.iloc[min_dd_idx]
        
        if mdd_date is not None:
            # 期間の文字列を作成（前回のリバランス日から今回のリバランス日まで）
            if isinstance(min_dd_idx, (int, np.integer)) and min_dd_idx > 0 and min_dd_idx < len(dates):
                prev_date = dates.iloc[min_dd_idx - 1]
                if isinstance(prev_date, pd.Timestamp) and isinstance(mdd_date, pd.Timestamp):
                    prev_year = prev_date.year - 1
                    curr_year = mdd_date.year
                    mdd_period = f"{prev_year}年10月〜{curr_year}年10月"
            elif isinstance(mdd_date, pd.Timestamp):
                prev_year = mdd_date.year - 1
                curr_year = mdd_date.year
                mdd_period = f"{prev_year}年10月〜{curr_year}年10月"
    
    return {
        'max_drawdown': max_dd_value,
        'mdd_date': mdd_date,
        'mdd_period': mdd_period
    }

def calculate_additional_metrics(results_df: pd.DataFrame) -> dict:
    """追加パフォーマンス指標の計算"""
    
    returns = results_df['strategy_return_net']
    dates = results_df.get('date', None)  # 日付列があれば使用
    
    # MDD計算（発生日も含む）
    mdd_result = calculate_max_drawdown(returns, dates)
    
    metrics = {
        'annual_return': calculate_annual_performance(returns, dates),
        'sharpe_ratio': calculate_sharpe_ratio(returns, risk_free_rate=0.0, dates=dates),
        'sortino_ratio': calculate_sortino_ratio(returns),
        'calmar_ratio': calculate_calmar_ratio(returns, dates),
        'max_drawdown': mdd_result['max_drawdown'],
        'mdd_date': mdd_result['mdd_date'],
        'mdd_period': mdd_result['mdd_period'],
        'win_rate': (returns > 0).sum() / len(returns) * 100,
        'avg_win': returns[returns > 0].mean() * 100 if (returns > 0).any() else 0,
        'avg_loss': returns[returns < 0].mean() * 100 if (returns < 0).any() else 0,
        'profit_factor': abs(returns[returns > 0].sum() / returns[returns < 0].sum()) if (returns < 0).any() else np.inf,
        'var_95': np.percentile(returns, 5) * 100,
        'cvar_95': returns[returns <= np.percentile(returns, 5)].mean() * 100,
        'volatility': calculate_volatility_annualized(returns, dates),
        'avg_annual_tax': results_df['tax'].mean() * 100,
        'total_tax': results_df['tax'].sum() * 100,
        'avg_investment_ratio': results_df['investment_ratio'].mean() * 100,
    }
    
    return metrics

# ===================================
# H. 結果出力（10月1日リバランス・100株単位版）
# ===================================

def print_performance_report_october_unit(results_df: pd.DataFrame, metrics: dict):
    """パフォーマンスレポート出力（10月1日リバランス・100株単位版）"""
    
    # デバッグ情報：リターンの分布を確認
    returns = results_df['strategy_return_net']
    print("\n" + "=" * 80)
    print("🔍 デバッグ情報: リターン分布")
    print("=" * 80)
    print(f"  リターン数: {len(returns)}")
    print(f"  正のリターン数: {(returns > 0).sum()}")
    print(f"  負のリターン数: {(returns < 0).sum()}")
    print(f"  ゼロリターン数: {(returns == 0).sum()}")
    if (returns < 0).any():
        print(f"  最小リターン: {returns.min()*100:.2f}%")
        print(f"  最大リターン: {returns.max()*100:.2f}%")
        print(f"  平均リターン: {returns.mean()*100:.2f}%")
        print(f"  標準偏差: {returns.std()*100:.2f}%")
    print("=" * 80)
    
    print("\n" + "=" * 80)
    print("📊 10月1日リバランス戦略パフォーマンス結果（100株単位制限）")
    print("=" * 80)
    
    gross_returns = results_df['strategy_return_gross']
    gross_dates = results_df.get('date', None)
    gross_annual = calculate_annual_performance(gross_returns, gross_dates)
    gross_sharpe = calculate_sharpe_ratio(gross_returns, risk_free_rate=0.0, dates=gross_dates)
    gross_mdd_result = calculate_max_drawdown(gross_returns, gross_dates)
    gross_dd = gross_mdd_result['max_drawdown']
    
    topix_annual = calculate_annual_performance(results_df['topix_return'])
    
    print(f"\n【戦略設定】")
    print(f"  ポジション:          ロングのみ（ショートなし）")
    print(f"  リバランス頻度:      年次（10月1日）")
    print(f"  制約:                100株単位購入制限")
    print(f"  平均ロング銘柄数:    {results_df['long_count'].mean():.1f}銘柄")
    print(f"  平均投資比率:        {metrics['avg_investment_ratio']:.1f}%")
    
    print(f"\n【税引き前パフォーマンス】")
    print(f"  年率リターン:        {gross_annual:8.2f}%")
    print(f"  シャープレシオ:      {gross_sharpe:8.2f}")
    print(f"  最大DD:              {gross_dd:8.2f}%")
    
    print(f"\n【税引き後パフォーマンス（実質）】")
    print(f"  年率リターン:        {metrics['annual_return']:8.2f}%")
    print(f"  シャープレシオ:      {metrics['sharpe_ratio']:8.2f}")
    # ソルティノレシオが無限大の場合は適切に表示
    sortino_display = metrics['sortino_ratio']
    if np.isinf(sortino_display):
        print(f"  ソルティノレシオ:    {'無限大':>8s} (損失なし)")
    else:
        print(f"  ソルティノレシオ:    {sortino_display:8.2f}")
    print(f"  カルマーレシオ:      {metrics['calmar_ratio']:8.2f}")
    print(f"  最大DD:              {metrics['max_drawdown']:8.2f}%")
    # MDD発生日を表示
    if metrics.get('mdd_date') is not None:
        mdd_date_str = metrics['mdd_date'].strftime('%Y年%m月%d日') if isinstance(metrics['mdd_date'], pd.Timestamp) else str(metrics['mdd_date'])
        if metrics.get('mdd_period'):
            print(f"  MDD発生日:          {mdd_date_str} ({metrics['mdd_period']})")
        else:
            print(f"  MDD発生日:          {mdd_date_str}")
    print(f"  年率ボラティリティ:  {metrics['volatility']:8.2f}%")
    
    print(f"\n【勝率・損益】")
    print(f"  年次勝率:            {metrics['win_rate']:8.1f}%")
    print(f"  平均利益（年次）:    {metrics['avg_win']:8.2f}%")
    print(f"  平均損失（年次）:    {metrics['avg_loss']:8.2f}%")
    print(f"  プロフィットファクター: {metrics['profit_factor']:8.2f}")
    
    print(f"\n【税金コストの影響】")
    print(f"  リターン減少:        {gross_annual - metrics['annual_return']:8.2f}%")
    print(f"  平均年次税負担:      {metrics['avg_annual_tax']:8.3f}%")
    print(f"  累積税負担:          {metrics['total_tax']:8.2f}%")
    
    print(f"\n【ベンチマーク比較】")
    print(f"  TOPIX年率リターン:   {topix_annual:8.2f}%")
    print(f"  税引き前 vs TOPIX:   {gross_annual - topix_annual:+8.2f}%")
    print(f"  税引き後 vs TOPIX:   {metrics['annual_return'] - topix_annual:+8.2f}%")
    
    print(f"\n【S&P500比較（参考年率12.0%）】")
    print(f"  税引き前 vs S&P500:  {gross_annual - 12.0:+8.2f}%")
    print(f"  税引き後 vs S&P500:  {metrics['annual_return'] - 12.0:+8.2f}%")
    
    # 年次パフォーマンステーブル
    print("\n" + "=" * 110)
    print("📅 年次パフォーマンス詳細（10月〜翌年10月）")
    print("=" * 110)
    
    print(f"\n{'期間':^20} {'税引前':>10} {'税引後':>10} {'TOPIX':>10} {'超過':>10} {'税負担':>8} {'銘柄数':>6} {'投資比率':>8}")
    print(f"{'':^20} {'リターン':>10} {'リターン':>10} {'リターン':>10} {'リターン':>10} {'(%)':>8} {'':>6} {'(%)':>8}")
    print("-" * 110)
    
    for _, row in results_df.iterrows():
        start_year = row['date'].year - 1
        end_year = row['date'].year
        period_str = f"{start_year}年10月〜{end_year}年10月"
        print(f"{period_str:^20} "
              f"{row['strategy_return_gross']*100:>9.2f}% "
              f"{row['strategy_return_net']*100:>9.2f}% "
              f"{row['topix_return']*100:>9.2f}% "
              f"{(row['strategy_return_net'] - row['topix_return'])*100:>+9.2f}% "
              f"{row['tax']*100:>7.3f}% "
              f"{int(row['long_count']):>6d} "
              f"{row['investment_ratio']*100:>7.1f}%")
    
    print("\n" + "=" * 110)

# ===================================
# I. メイン実行
# ===================================

if __name__ == "__main__":
    
    print("=" * 80)
    print("🏆 SMBC「割安高質」戦略 - 10月1日リバランス版（100株単位制限）")
    print("=" * 80)
    print(f"\n📋 検証概要:")
    print(f"  戦略名: 割安高質（ロングオンリー）")
    print(f"  検証期間: 2016年10月〜2025年10月")
    print(f"  リバランス: 年次（10月1日）")
    print(f"  ポジション: ロングのみ（ショートなし）")
    print(f"  制約: 100株単位購入制限")
    print(f"  譲渡益税: {TAX_RATE*100:.3f}%")
    print(f"  実行日時: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    print("=" * 80)
    
    # データ読み込み
    statements_df = load_financial_data()
    if statements_df.empty:
        logger.error("財務データ読み込み失敗")
        exit(1)
    
    prices_df = load_existing_price_data()
    if prices_df.empty:
        logger.error("株価データ読み込み失敗")
        exit(1)
    
    # 財務指標計算
    enhanced_financial_data = calculate_market_metrics_fast_chunked(statements_df, prices_df, chunk_size=200)
    if enhanced_financial_data.empty:
        logger.error("財務指標計算失敗")
        exit(1)
    
    # バックテスト実行
    results = smbc_value_quality_backtest_october_unit(enhanced_financial_data, prices_df)
    if results.empty:
        logger.error("バックテスト実行失敗")
        exit(1)
    
    # パフォーマンス指標計算
    metrics = calculate_additional_metrics(results)
    
    # 結果出力
    print_performance_report_october_unit(results, metrics)
    
    # 結果保存
    results.to_csv('backtest_results_october_unit.csv', index=False, encoding='utf-8-sig')
    
    logger.info("結果をCSVに保存: backtest_results_october_unit.csv")
    
    print("\n" + "=" * 80)
    print("🏆 10月1日リバランス戦略検証完了（100株単位制限版）")
    print("=" * 80)
    print(f"\n📁 出力ファイル:")
    print(f"  - バックテスト結果: backtest_results_october_unit.csv")
    print(f"  - ログファイル: backtest_long_only_october_unit_constraints.log")
    print("\n")
```

```python
# ===================================
# 日次MDD計算関数（高速化版）
# ===================================

def calculate_max_drawdown_daily(
    results_df: pd.DataFrame,
    prices_df: pd.DataFrame,
    enhanced_financial_data: pd.DataFrame,
    initial_capital: float = 10_000_000,
    leverage: float = 1.0,
    interest_rate: float = 0.0198
) -> dict:
    """
    年次リバランス戦略の日次MDD計算（高速化版）
    
    Args:
        results_df: 年次リバランスの結果（date, strategy_return_net等を含む）
        prices_df: 株価データ
        enhanced_financial_data: 財務データ
        initial_capital: 初期資本
        leverage: レバレッジ倍率（デフォルト1.0倍）
        interest_rate: 信用金利（年率、デフォルト1.98%）
    
    Returns:
        dict: {
            'max_drawdown_daily': 日次MDD（%）,
            'daily_equity_curve': 日次時価総額のDataFrame,
            'drawdown_series': 日次ドローダウンのSeries
        }
    """
    
    logger.info("日次MDD計算を開始（高速化版）...")
    
    # リバランス日を取得
    rebalance_dates = sorted(results_df['date'].tolist())
    
    # 最初のリバランス日を取得（結果の最初のdateの前年10月1日）
    if len(rebalance_dates) == 0:
        logger.error("リバランス日が見つかりません")
        return {
            'max_drawdown_daily': 0.0,
            'daily_equity_curve': pd.DataFrame(),
            'drawdown_series': pd.Series()
        }
    
    first_date = rebalance_dates[0]
    start_year = first_date.year - 1
    start_date = pd.Timestamp(f'{start_year}-10-01')
    end_date = rebalance_dates[-1]
    
    # 価格データを事前に辞書化（高速アクセスのため）
    logger.info("価格データを辞書化中...")
    price_dict = {}
    period_prices = prices_df[
        (prices_df['Date'] >= start_date) & 
        (prices_df['Date'] <= end_date)
    ].copy()
    
    for _, row in tqdm(period_prices.iterrows(), total=len(period_prices), desc="価格データ辞書化", leave=False):
        code = str(row['Code'])
        date = row['Date']
        price = row['Close']
        
        if code not in price_dict:
            price_dict[code] = {}
        price_dict[code][date] = price
    
    logger.info(f"価格データ辞書化完了: {len(price_dict)}銘柄")
    
    # 全期間の営業日を取得
    all_dates = sorted(period_prices['Date'].unique())
    logger.info(f"計算対象日数: {len(all_dates)}日")
    
    # 日次時価総額を計算
    daily_equity = []
    current_capital = initial_capital
    
    # 各リバランス期間を処理
    for i, rebalance_date in enumerate(tqdm(rebalance_dates, desc="リバランス期間処理")):
        if i == 0:
            period_start = start_date
        else:
            period_start = rebalance_dates[i-1]
        
        period_end = rebalance_date
        
        # この期間の財務データを取得
        current_data = enhanced_financial_data[
            enhanced_financial_data['DisclosedDate'] <= period_start
        ].copy()
        current_data = current_data.sort_values('DisclosedDate').groupby('Code').tail(1)
        
        if len(current_data) < 100:
            # データ不足の場合は前回のポートフォリオを維持（簡略化）
            if i > 0 and daily_equity:
                # 前回の最終値を維持
                last_equity = daily_equity[-1]['equity']
                period_dates = [d for d in all_dates if period_start < d <= period_end]
                for date in period_dates:
                    daily_equity.append({
                        'date': date,
                        'equity': last_equity,
                        'period': i
                    })
            continue
        
        # PBRとROEでランキング
        current_data['PBR_Rank'] = current_data['PBR'].rank(method='first', ascending=True)
        current_data['ROE_Rank'] = current_data['ROE'].rank(method='first', ascending=False)
        
        # 四分位分割
        current_data['PBR_Quartile'] = pd.qcut(current_data['PBR_Rank'], q=4, labels=[1, 2, 3, 4], duplicates='drop')
        current_data['ROE_Quartile'] = pd.qcut(current_data['ROE_Rank'], q=4, labels=[1, 2, 3, 4], duplicates='drop')
        
        # ロング候補：低PBR × 高ROE
        long_candidates = current_data[
            (current_data['PBR_Quartile'] == 1) & 
            (current_data['ROE_Quartile'] == 4)
        ].nsmallest(50, 'PBR')
        
        # ポートフォリオ構築
        portfolio = build_unit_share_portfolio(
            long_candidates,
            target_positions=20,
            initial_capital=current_capital
        )
        
        if not portfolio['stocks']:
            continue
        
        # レバレッジを考慮した株数と借入額を事前計算
        base_investment = sum(portfolio['amounts'])
        if leverage > 1.0:
            leveraged_shares_list = []
            for base_shares in portfolio['shares']:
                leveraged_shares_list.append(int(base_shares * leverage / 100) * 100)
            actual_borrowed = base_investment * leverage - current_capital
        else:
            leveraged_shares_list = portfolio['shares']
            actual_borrowed = 0.0
        
        # この期間の日次価格データを取得
        period_dates = [d for d in all_dates if period_start < d <= period_end]
        
        # 日次計算（バッチ処理で高速化）
        for date in tqdm(period_dates, desc=f"期間{i+1}日次計算", leave=False):
            portfolio_value = 0.0
            
            # 各銘柄の時価を計算（辞書から高速取得）
            for j, code in enumerate(portfolio['stocks']):
                code_str = str(code)
                shares = leveraged_shares_list[j] if leverage > 1.0 else portfolio['shares'][j]
                
                # 辞書から価格を取得
                if code_str in price_dict and date in price_dict[code_str]:
                    price = price_dict[code_str][date]
                    portfolio_value += shares * price
            
            # レバレッジを考慮
            if leverage > 1.0:
                # 信用金利コストを計算
                period_years = (date - period_start).days / 365.25
                interest_cost = actual_borrowed * interest_rate * period_years
                portfolio_value -= interest_cost
            
            # 現金残高を追加
            if leverage == 1.0:
                cash_remaining = current_capital - base_investment
            else:
                cash_remaining = current_capital - base_investment * leverage
            
            total_value = portfolio_value + cash_remaining
            
            daily_equity.append({
                'date': date,
                'equity': total_value,
                'period': i
            })
        
        # 期間終了時の資本を更新
        if i < len(results_df):
            current_capital = current_capital * (1 + results_df.iloc[i]['strategy_return_net'])
    
    if not daily_equity:
        logger.warning("日次時価総額データが生成されませんでした")
        return {
            'max_drawdown_daily': 0.0,
            'daily_equity_curve': pd.DataFrame(),
            'drawdown_series': pd.Series()
        }
    
    logger.info("日次時価総額データを整理中...")
    
    # DataFrameに変換
    equity_df = pd.DataFrame(daily_equity)
    equity_df = equity_df.sort_values('date')
    equity_df = equity_df.drop_duplicates(subset=['date'], keep='last')
    
    # 累積最大値を計算
    equity_df['running_max'] = equity_df['equity'].cummax()
    
    # ドローダウンを計算
    equity_df['drawdown'] = (equity_df['equity'] - equity_df['running_max']) / equity_df['running_max'] * 100
    
    # 最大ドローダウンと発生日を取得
    max_dd = equity_df['drawdown'].min()
    mdd_row = equity_df.loc[equity_df['drawdown'].idxmin()]
    mdd_date = mdd_row['date']
    
    logger.info(f"日次MDD計算完了: {max_dd:.2f}% (発生日: {mdd_date.strftime('%Y年%m月%d日') if isinstance(mdd_date, pd.Timestamp) else mdd_date})")
    
    return {
        'max_drawdown_daily': max_dd,
        'mdd_date': mdd_date,
        'daily_equity_curve': equity_df[['date', 'equity', 'drawdown']],
        'drawdown_series': equity_df.set_index('date')['drawdown']
    }
```

```python
import pandas as pd
import numpy as np
from tqdm import tqdm 
import warnings
import os
import logging
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

warnings.filterwarnings('ignore')

# ===================================
# ロギング設定
# ===================================
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[
        logging.FileHandler('backtest_daily_value_quality.log', encoding='utf-8'),
        logging.StreamHandler()
    ]
)
logger = logging.getLogger(__name__)

# ===================================
# グローバル変数
# ===================================
CACHE_FILE = 'topix_quarterly_statements.csv'
PRICE_CACHE = {}

# 税金パラメータ
TAX_RATE = 0.20315  # 譲渡益税 20.315%

# 単位株制限
UNIT_SHARES = 100  # 100株単位

# ===================================
# A. 財務データ読み込み（年次版と同じ）
# ===================================

def load_financial_data(cache_filename: str = CACHE_FILE) -> pd.DataFrame:
    """財務諸表データをキャッシュから読み込み（列名を自動判定）"""
    
    if not os.path.exists(cache_filename):
        logger.error(f"財務データファイル '{cache_filename}' が見つかりません")
        return pd.DataFrame()
    
    try:
        df = pd.read_csv(
            cache_filename, 
            encoding='utf-8-sig',
            parse_dates=['DisclosedDate'],
            low_memory=False
        )
        
        logger.info(f"財務データ読み込み成功: {len(df):,}件")
        
        # 列名のマッピング
        column_mapping = {
            'IssuedShareTotal': ['IssuedShareTotal', 'NumberOfIssuedAndOutstandingSharesAtTheEndOfFiscalYearIncludingTreasuryStock'],
            'Equity': ['Equity', 'NetAssets', 'TotalEquity'],
            'Profit': ['Profit', 'NetIncome', 'ProfitAttributableToOwnersOfParent']
        }
        
        available_cols = df.columns.tolist()
        required_columns = {}
        
        for target_col, possible_names in column_mapping.items():
            found = False
            for possible_name in possible_names:
                if possible_name in available_cols:
                    required_columns[target_col] = possible_name
                    found = True
                    break
            if not found:
                required_columns[target_col] = None
        
        rename_dict = {v: k for k, v in required_columns.items() if v is not None}
        df = df.rename(columns=rename_dict)
        
        for col in ['IssuedShareTotal', 'Equity', 'Profit']:
            if col not in df.columns:
                df[col] = 0
        
        base_cols = ['Code', 'DisclosedDate']
        if 'CompanyName' in df.columns:
            base_cols.append('CompanyName')
        
        final_cols = base_cols + ['Profit', 'Equity', 'IssuedShareTotal']
        df = df[final_cols].copy()
        
        logger.info(f"使用列: {df.columns.tolist()}")
        
        return df
    
    except Exception as e:
        logger.error(f"財務データ読み込みエラー: {e}", exc_info=True)
        return pd.DataFrame()

# ===================================
# B. 株価データ読み込み（年次版と同じ）
# ===================================

def load_existing_price_data(ohlcv_dir: str = './OHLCV_Adjusted') -> pd.DataFrame:
    """株価データ読み込み（メモリ効率化版）"""
    global PRICE_CACHE
    
    if not os.path.exists(ohlcv_dir):
        logger.error(f"株価データディレクトリ '{ohlcv_dir}' が見つかりません。")
        return pd.DataFrame()
    
    csv_files = sorted([
        f for f in os.listdir(ohlcv_dir) 
        if f.startswith('OHLCV_Adjusted_') and f.endswith('.csv') and f != 'OHLCV_Adjusted_TOPIX.csv'
    ])
    
    if not csv_files:
        logger.error(f"ディレクトリ '{ohlcv_dir}' 内にCSVファイルが見つかりません。")
        return pd.DataFrame()
    
    logger.info(f"株価ファイル数: {len(csv_files)}個")
    
    all_dataframes = []
    usecols = ['Date', 'Ticker', 'AdjustmentClose']
    
    for csv_file in tqdm(csv_files, desc="株価ファイル読み込み中"):
        file_path = os.path.join(ohlcv_dir, csv_file)
        try:
            df = pd.read_csv(
                file_path,
                usecols=usecols,
                parse_dates=['Date'],
                dtype={'Ticker': 'Int64', 'AdjustmentClose': 'float32'}
            )
            
            df = df.drop_duplicates(subset=['Ticker', 'Date'], keep='first')
            all_dataframes.append(df)
            
        except Exception as e:
            logger.warning(f"ファイル読み込みエラー ({csv_file}): {e}")
            continue
    
    if not all_dataframes:
        return pd.DataFrame()
    
    logger.info("全ファイルを結合中...")
    df_all = pd.concat(all_dataframes, ignore_index=True)
    
    logger.info(f"結合完了: {len(df_all):,}件")
    
    df_all = df_all.rename(columns={'Ticker': 'Code', 'AdjustmentClose': 'Close'})
    df_all['Code'] = df_all['Code'].astype('str').str.replace('<NA>', '0').str.zfill(4)
    df_all = df_all.dropna(subset=['Close'])
    
    logger.info("重複除去 & ソート中...")
    df_all = df_all.sort_values(['Code', 'Date'])
    df_all = df_all.drop_duplicates(subset=['Code', 'Date'], keep='first')
    
    logger.info(f"重複除去後: {len(df_all):,}件")
    
    return df_all

# ===================================
# C. 財務指標計算（年次版と同じ）
# ===================================

def safe_code_to_int(code_series: pd.Series) -> pd.Series:
    """証券コードを整数に変換（ゼロ埋め対応）"""
    cleaned = code_series.astype(str).str.replace(r'\D', '', regex=True)
    cleaned = cleaned.replace('', '0')
    return pd.to_numeric(cleaned, errors='coerce').fillna(0).astype('int64')

def calculate_market_metrics_fast_chunked(statements_df: pd.DataFrame, 
                                          prices_df: pd.DataFrame,
                                          chunk_size: int = 200) -> pd.DataFrame:
    """時価総額・PBR・ROE計算（チャンク処理版）"""
    
    logger.info("時価総額・PBR・ROE計算中（チャンク処理版）...")
    
    statements_df = statements_df.copy()
    statements_df['Profit'] = pd.to_numeric(statements_df['Profit'], errors='coerce').fillna(0)
    statements_df['Equity'] = pd.to_numeric(statements_df['Equity'], errors='coerce').fillna(0)
    statements_df['IssuedShareTotal'] = pd.to_numeric(statements_df['IssuedShareTotal'], errors='coerce').fillna(1)
    
    statements_df = statements_df[
        (statements_df['Equity'] > 0) & 
        (statements_df['IssuedShareTotal'] > 0)
    ]
    logger.info(f"有効な財務データ: {len(statements_df):,}件")
    
    # Code変換
    logger.info("Code列を整数型に変換中...")
    statements_df['Code_int'] = safe_code_to_int(statements_df['Code'])
    prices_df['Code_int'] = safe_code_to_int(prices_df['Code'])
    
    statements_df = statements_df[statements_df['Code_int'] > 0]
    prices_df = prices_df[prices_df['Code_int'] > 0]
    
    # ソート
    statements_df = statements_df.sort_values(['Code_int', 'DisclosedDate']).reset_index(drop=True)
    prices_df = prices_df.sort_values(['Code_int', 'Date']).reset_index(drop=True)
    
    # 重複除去
    statements_df = statements_df.drop_duplicates(subset=['Code_int', 'DisclosedDate'], keep='first')
    prices_df = prices_df.drop_duplicates(subset=['Code_int', 'Date'], keep='first')
    
    logger.info(f"ソート・重複除去後: 財務 {len(statements_df):,}件, 株価 {len(prices_df):,}件")
    
    # 銘柄ごとにマージ
    logger.info("銘柄ごとにマージ中...")
    
    statements_groups = list(statements_df.groupby('Code_int'))
    prices_dict = {code: group for code, group in prices_df.groupby('Code_int')}
    
    merged_list = []
    num_chunks = (len(statements_groups) + chunk_size - 1) // chunk_size
    
    for chunk_idx in tqdm(range(num_chunks), desc="マージ処理"):
        start_idx = chunk_idx * chunk_size
        end_idx = min((chunk_idx + 1) * chunk_size, len(statements_groups))
        
        chunk_groups = statements_groups[start_idx:end_idx]
        
        for code, stmt_code in chunk_groups:
            if code not in prices_dict:
                continue
            
            price_code = prices_dict[code]
            
            if len(price_code) == 0:
                continue
            
            stmt_code = stmt_code.sort_values('DisclosedDate').reset_index(drop=True)
            price_code = price_code.sort_values('Date').reset_index(drop=True)
            
            try:
                merged = pd.merge_asof(
                    stmt_code,
                    price_code[['Date', 'Close']],
                    left_on='DisclosedDate',
                    right_on='Date',
                    direction='backward',
                    tolerance=pd.Timedelta(days=10)
                )
                
                if not merged.empty:
                    merged_list.append(merged)
            except Exception:
                continue
    
    if not merged_list:
        logger.error("マージ結果が空です")
        return pd.DataFrame()
    
    df_merged = pd.concat(merged_list, ignore_index=True)
    df_merged = df_merged.dropna(subset=['Close'])
    
    logger.info(f"マージ完了: {len(df_merged):,}件")
    
    # 財務指標計算
    df_merged['MarketCap'] = df_merged['Close'] * df_merged['IssuedShareTotal']
    df_merged['PBR'] = df_merged['MarketCap'] / df_merged['Equity']
    df_merged['ROE'] = (df_merged['Profit'] / df_merged['Equity']) * 100
    
    result_cols = ['Code', 'DisclosedDate', 'Close', 'MarketCap', 'PBR', 'ROE', 'Date']
    
    if 'CompanyName' in df_merged.columns:
        result_cols.insert(1, 'CompanyName')
    
    result_df = df_merged[result_cols].copy()
    result_df = result_df.rename(columns={'Close': 'StockPrice', 'Date': 'PriceDate'})
    
    # 異常値除去
    mask = (
        (result_df['PBR'] > 0) &
        (result_df['PBR'] < 50) &
        (result_df['ROE'] > -100) &
        (result_df['ROE'] < 100) &
        (result_df['MarketCap'] > 1_000_000_000)
    )
    
    result_df = result_df[mask].copy()
    
    logger.info(f"計算完了: {len(result_df):,}件")
    
    return result_df

# ===================================
# D. ポートフォリオ構築（年次版と同じ）
# ===================================

def build_unit_share_portfolio(stock_candidates: pd.DataFrame,
                               target_positions: int = 20,
                               initial_capital: float = 10_000_000) -> dict:
    """100株単位ポートフォリオ構築"""
    
    if len(stock_candidates) == 0:
        return {
            'stocks': [],
            'shares': [],
            'prices': [],
            'amounts': []
        }
    
    # 上位N銘柄を選択
    selected = stock_candidates.head(target_positions).copy()
    
    # 1銘柄あたりの配分資金
    capital_per_stock = initial_capital / len(selected)
    
    stocks = []
    shares_list = []
    prices_list = []
    amounts_list = []
    
    for _, row in selected.iterrows():
        code = row['Code']
        price = row['StockPrice']
        
        # 100株単位で購入可能な株数を計算
        required_amount = price * UNIT_SHARES
        
        if required_amount <= capital_per_stock:
            shares = int(capital_per_stock // required_amount) * UNIT_SHARES
            
            if shares > 0:
                stocks.append(code)
                shares_list.append(shares)
                prices_list.append(price)
                amounts_list.append(shares * price)
    
    return {
        'stocks': stocks,
        'shares': shares_list,
        'prices': prices_list,
        'amounts': amounts_list
    }

# ===================================
# E. 日次リターン計算（高速化版）
# ===================================

def preprocess_price_data(prices_df: pd.DataFrame) -> dict:
    """価格データを高速検索用の辞書に変換"""
    
    logger.info("価格データを辞書化中...")
    
    # Code × Date の辞書を作成
    price_dict = {}
    
    for code, group in tqdm(prices_df.groupby('Code'), desc="価格辞書化"):
        group = group.sort_values('Date')
        price_dict[code] = {
            'dates': group['Date'].values,
            'prices': group['Close'].values
        }
    
    logger.info(f"価格辞書化完了: {len(price_dict)}銘柄")
    
    return price_dict

def get_price_fast(code: str, target_date: pd.Timestamp, price_dict: dict, tolerance_days: int = 2) -> float:
    """高速価格取得（辞書ベース）"""
    
    if code not in price_dict:
        return np.nan
    
    dates = price_dict[code]['dates']
    prices = price_dict[code]['prices']
    
    # 二分探索で最も近い日付を探す
    # datesはpd.Timestampの配列なので、target_dateと直接比較可能
    target_ts = target_date.value  # ナノ秒単位のTimestampに変換
    
    # tolerance内の価格を探す
    tolerance_ns = tolerance_days * 24 * 60 * 60 * 10**9
    
    # datesをナノ秒単位の整数配列に変換
    dates_ns = dates.astype('datetime64[ns]').astype('int64')
    
    # 最も近い日付のインデックスを探す
    idx = np.searchsorted(dates_ns, target_ts)
    
    # 前後の日付をチェック
    candidates = []
    
    if idx < len(dates):
        diff = abs(dates_ns[idx] - target_ts)
        candidates.append((diff, prices[idx]))
    
    if idx > 0:
        diff = abs(dates_ns[idx-1] - target_ts)
        candidates.append((diff, prices[idx-1]))
    
    if not candidates:
        return np.nan
    
    # 最も近い価格を返す
    min_diff, best_price = min(candidates, key=lambda x: x[0])
    
    if min_diff <= tolerance_ns:
        return best_price
    else:
        return np.nan

def calculate_daily_return_fast(
    portfolio: dict,
    prev_date: pd.Timestamp,
    current_date: pd.Timestamp,
    price_dict: dict,
    initial_capital: float
) -> dict:
    """日次ポートフォリオリターン計算（高速版）"""
    
    if not portfolio['stocks']:
        return {
            'return': 0.0,
            'value': initial_capital,
            'cash': initial_capital
        }
    
    stocks = portfolio['stocks']
    shares = portfolio['shares']
    
    # 各銘柄の損益計算（ベクトル化）
    prev_value = 0.0
    curr_value = 0.0
    
    for i, code in enumerate(stocks):
        prev_price = get_price_fast(code, prev_date, price_dict)
        curr_price = get_price_fast(code, current_date, price_dict)
        
        if np.isnan(prev_price) or np.isnan(curr_price):
            continue
        
        share_count = shares[i]
        
        prev_value += share_count * prev_price
        curr_value += share_count * curr_price
    
    if prev_value == 0:
        return {
            'return': 0.0,
            'value': initial_capital,
            'cash': initial_capital
        }
    
    daily_return = (curr_value - prev_value) / prev_value
    
    return {
        'return': daily_return,
        'value': curr_value,
        'cash': initial_capital - sum(portfolio['amounts'])
    }

# ===================================
# F. 日次バックテスト（高速化版）
# ===================================

def smbc_value_quality_backtest_daily_fast(enhanced_financial_data: pd.DataFrame, 
                                          prices_df: pd.DataFrame,
                                          initial_capital: float = 10_000_000,
                                          rebalance_frequency: int = 1) -> pd.DataFrame:
    """
    SMBC割安高質戦略（日次リバランス・高速化版）
    """
    
    logger.info(f"日次リバランス戦略バックテスト実行中（高速化版）...")
    
    # 価格データを辞書化（事前処理）
    price_dict = preprocess_price_data(prices_df)
    
    # 全営業日を取得
    all_dates = pd.to_datetime(prices_df['Date'].unique())
    all_dates = pd.DatetimeIndex(sorted(all_dates))
    
    # バックテスト期間
    start_date = pd.Timestamp('2016-10-01')
    end_date = pd.Timestamp('2025-10-01')
    
    trading_dates = all_dates[(all_dates >= start_date) & (all_dates <= end_date)]
    
    logger.info(f"バックテスト期間: {start_date.date()} 〜 {end_date.date()}")
    logger.info(f"営業日数: {len(trading_dates)}日")
    logger.info(f"リバランス頻度: {rebalance_frequency}営業日ごと")
    
    # 財務データのキャッシュ（日付でソート）
    financial_cache = enhanced_financial_data.sort_values('DisclosedDate').copy()
    
    # 初期化
    current_portfolio = {
        'stocks': [],
        'shares': [],
        'prices': [],
        'amounts': []
    }
    
    cumulative_capital = initial_capital
    last_rebalance_date = None
    rebalance_counter = 0
    
    strategy_results = []
    
    # 日次ループ（高速化版）
    for date_idx, current_date in enumerate(tqdm(trading_dates, desc="日次バックテスト（高速化版）")):
        
        # リバランス判定
        should_rebalance = False
        
        if last_rebalance_date is None:
            should_rebalance = True
        elif rebalance_counter >= rebalance_frequency:
            should_rebalance = True
        
        # リバランス実行
        if should_rebalance:
            # 現在の財務データを取得（高速化：事前ソート済み）
            current_data = financial_cache[financial_cache['DisclosedDate'] <= current_date]
            
            if len(current_data) > 0:
                current_data = current_data.groupby('Code', as_index=False).last()
            
            if len(current_data) >= 100:
                # PBRとROEでランキング
                current_data['PBR_Rank'] = current_data['PBR'].rank(method='first', ascending=True)
                current_data['ROE_Rank'] = current_data['ROE'].rank(method='first', ascending=False)
                
                # 四分位分割
                current_data['PBR_Quartile'] = pd.qcut(current_data['PBR_Rank'], q=4, labels=[1, 2, 3, 4], duplicates='drop')
                current_data['ROE_Quartile'] = pd.qcut(current_data['ROE_Rank'], q=4, labels=[1, 2, 3, 4], duplicates='drop')
                
                # ロング候補：低PBR × 高ROE
                long_candidates = current_data[
                    (current_data['PBR_Quartile'] == 1) & 
                    (current_data['ROE_Quartile'] == 4)
                ].nsmallest(50, 'PBR')
                
                # ポートフォリオ構築
                current_portfolio = build_unit_share_portfolio(
                    long_candidates,
                    target_positions=20,
                    initial_capital=cumulative_capital
                )
                
                last_rebalance_date = current_date
                rebalance_counter = 0
        
        # 日次リターン計算（高速版）
        if date_idx > 0:
            prev_date = trading_dates[date_idx - 1]
            
            daily_result = calculate_daily_return_fast(
                current_portfolio,
                prev_date,
                current_date,
                price_dict,
                cumulative_capital
            )
            
            daily_return = daily_result['return']
            
            # 累積資本を更新
            cumulative_capital = cumulative_capital * (1 + daily_return)
            
            # 結果記録
            strategy_results.append({
                'date': current_date,
                'daily_return': daily_return,
                'cumulative_capital': cumulative_capital,
                'portfolio_value': daily_result['value'],
                'num_positions': len(current_portfolio['stocks']),
                'is_rebalance_day': should_rebalance
            })
        
        rebalance_counter += 1
    
    return pd.DataFrame(strategy_results)

# ===================================
# G. パフォーマンス分析（年次版を流用）
# ===================================

def calculate_annual_performance(returns: pd.Series, dates: pd.Series = None) -> float:
    """年率リターン計算"""
    if len(returns) == 0:
        return 0.0
    
    cumulative = (1 + returns).prod()
    
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
    else:
        years = len(returns) / 252  # 営業日ベース
    
    if years > 0:
        return ((cumulative ** (1/years)) - 1) * 100
    else:
        return 0.0

def calculate_volatility_annualized(returns: pd.Series, dates: pd.Series = None) -> float:
    """年率換算ボラティリティ計算"""
    if len(returns) == 0:
        return 0.0
    
    period_std = returns.std()
    
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
        periods_per_year = len(returns) / years if years > 0 else 252
    else:
        periods_per_year = 252  # 営業日ベース
    
    if periods_per_year > 0:
        return period_std * np.sqrt(periods_per_year) * 100
    else:
        return 0.0

def calculate_sharpe_ratio(returns: pd.Series, risk_free_rate: float = 0.0, dates: pd.Series = None) -> float:
    """シャープレシオ計算（年率換算）"""
    if len(returns) == 0 or returns.std() == 0:
        return 0.0
    
    excess_returns = returns - risk_free_rate / 252  # 日次無リスク金利
    mean_return = excess_returns.mean()
    std_return = excess_returns.std()
    
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
        periods_per_year = len(returns) / years if years > 0 else 252
    else:
        periods_per_year = 252
    
    if std_return > 0:
        return (mean_return / std_return) * np.sqrt(periods_per_year)
    else:
        return 0.0

def calculate_max_drawdown(returns: pd.Series, dates: pd.Series = None) -> dict:
    """最大ドローダウン計算"""
    if len(returns) == 0:
        return {
            'max_drawdown': 0.0,
            'mdd_date': None,
            'mdd_period': None
        }
    
    cumulative = (1 + returns).cumprod()
    running_max = cumulative.cummax()
    drawdown = (cumulative - running_max) / running_max
    
    min_dd_idx = drawdown.idxmin()
    max_dd_value = drawdown.min() * 100
    
    mdd_date = None
    if dates is not None and len(dates) > 0:
        if min_dd_idx in dates.index:
            mdd_date = dates.loc[min_dd_idx]
        elif isinstance(min_dd_idx, (int, np.integer)) and 0 <= min_dd_idx < len(dates):
            mdd_date = dates.iloc[min_dd_idx]
    
    return {
        'max_drawdown': max_dd_value,
        'mdd_date': mdd_date,
        'mdd_period': None
    }

def calculate_additional_metrics(results_df: pd.DataFrame) -> dict:
    """追加パフォーマンス指標の計算"""
    
    returns = results_df['daily_return']
    dates = results_df.get('date', None)
    
    # MDD計算
    mdd_result = calculate_max_drawdown(returns, dates)
    
    metrics = {
        'annual_return': calculate_annual_performance(returns, dates),
        'sharpe_ratio': calculate_sharpe_ratio(returns, risk_free_rate=0.0, dates=dates),
        'max_drawdown': mdd_result['max_drawdown'],
        'mdd_date': mdd_result['mdd_date'],
        'win_rate': (returns > 0).sum() / len(returns) * 100,
        'avg_win': returns[returns > 0].mean() * 100 if (returns > 0).any() else 0,
        'avg_loss': returns[returns < 0].mean() * 100 if (returns < 0).any() else 0,
        'volatility': calculate_volatility_annualized(returns, dates),
        'final_capital': results_df['cumulative_capital'].iloc[-1],
        'total_return': (results_df['cumulative_capital'].iloc[-1] / results_df['cumulative_capital'].iloc[0] - 1) * 100,
    }
    
    return metrics

# ===================================
# H. 結果出力（日次版）
# ===================================

def print_performance_report_daily(results_df: pd.DataFrame, metrics: dict, initial_capital: float):
    """パフォーマンスレポート出力（日次版）"""
    
    print("\n" + "=" * 80)
    print("📊 日次リバランス戦略パフォーマンス結果（100株単位制限）")
    print("=" * 80)
    
    print(f"\n【戦略設定】")
    print(f"  ポジション:          ロングのみ（ショートなし）")
    print(f"  リバランス頻度:      日次")
    print(f"  制約:                100株単位購入制限")
    print(f"  平均ポジション数:    {results_df['num_positions'].mean():.1f}銘柄")
    print(f"  初期資本:            ¥{initial_capital:,.0f}")
    
    print(f"\n【パフォーマンス】")
    print(f"  最終資本:            ¥{metrics['final_capital']:,.0f}")
    print(f"  累積リターン:        {metrics['total_return']:.2f}%")
    print(f"  年率リターン:        {metrics['annual_return']:.2f}%")
    print(f"  シャープレシオ:      {metrics['sharpe_ratio']:.2f}")
    print(f"  最大DD:              {metrics['max_drawdown']:.2f}%")
    if metrics.get('mdd_date'):
        mdd_date_str = metrics['mdd_date'].strftime('%Y年%m月%d日')
        print(f"  MDD発生日:          {mdd_date_str}")
    print(f"  年率ボラティリティ:  {metrics['volatility']:.2f}%")
    
    print(f"\n【勝率・損益（日次）】")
    print(f"  勝率:                {metrics['win_rate']:.1f}%")
    print(f"  平均利益:            {metrics['avg_win']:.3f}%")
    print(f"  平均損失:            {metrics['avg_loss']:.3f}%")
    
    print(f"\n【年次版との比較参考】")
    print(f"  年次版年率リターン:  28.52%")
    print(f"  日次版年率リターン:  {metrics['annual_return']:.2f}%")
    print(f"  差分:                {metrics['annual_return'] - 28.52:+.2f}%")
    
    print("\n" + "=" * 80)

# ===================================
# I. メイン実行
# ===================================

if __name__ == "__main__":
    
    print("=" * 80)
    print("🚀 SMBC「割安高質」戦略 - 日次リバランス版（高速化）")
    print("=" * 80)
    print(f"\n📋 検証概要:")
    print(f"  戦略名: 割安高質（ロングオンリー）")
    print(f"  検証期間: 2016年10月〜2025年10月")
    print(f"  リバランス: 日次")
    print(f"  ポジション: ロングのみ（ショートなし）")
    print(f"  制約: 100株単位購入制限")
    print(f"  最適化: 価格辞書化 + NumPy高速化")
    print(f"  実行日時: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    print("=" * 80)
    
    # データ読み込み
    statements_df = load_financial_data()
    if statements_df.empty:
        logger.error("財務データ読み込み失敗")
        exit(1)
    
    prices_df = load_existing_price_data()
    if prices_df.empty:
        logger.error("株価データ読み込み失敗")
        exit(1)
    
    # 財務指標計算
    enhanced_financial_data = calculate_market_metrics_fast_chunked(statements_df, prices_df, chunk_size=200)
    if enhanced_financial_data.empty:
        logger.error("財務指標計算失敗")
        exit(1)
    
    # 日次バックテスト実行（高速化版）
    results = smbc_value_quality_backtest_daily_fast(enhanced_financial_data, prices_df, rebalance_frequency=1)
    if results.empty:
        logger.error("バックテスト実行失敗")
        exit(1)
    
    # パフォーマンス指標計算
    metrics = calculate_additional_metrics(results)
    
    # 結果出力
    print_performance_report_daily(results, metrics, 10_000_000)
    
    # 結果保存
    results.to_csv('backtest_results_daily_fast.csv', index=False, encoding='utf-8-sig')
    
    logger.info("結果をCSVに保存: backtest_results_daily_fast.csv")
    
    print("\n" + "=" * 80)
    print("🏆 日次リバランス戦略検証完了（高速化版）")
    print("=" * 80)
    print(f"\n📁 出力ファイル:")
    print(f"  - バックテスト結果: backtest_results_daily_fast.csv")
    print(f"  - ログファイル: backtest_daily_value_quality.log")
    print("\n")
```

```python
import pandas as pd
import numpy as np
from tqdm import tqdm 
import warnings
import os
import logging
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

warnings.filterwarnings('ignore')

# ===================================
# ロギング設定
# ===================================
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[
        logging.FileHandler('backtest_daily_value_quality_complete.log', encoding='utf-8'),
        logging.StreamHandler()
    ]
)
logger = logging.getLogger(__name__)

# ===================================
# グローバル変数
# ===================================
CACHE_FILE = 'topix_quarterly_statements.csv'
PRICE_CACHE = {}

# 税金パラメータ
TAX_RATE = 0.20315  # 譲渡益税 20.315%

# 単位株制限
UNIT_SHARES = 100  # 100株単位

# スリッページ
SLIPPAGE = 0.003  # 0.3%

# ===================================
# A. 財務データ読み込み
# ===================================

def load_financial_data(cache_filename: str = CACHE_FILE) -> pd.DataFrame:
    """財務諸表データをキャッシュから読み込み（列名を自動判定）"""
    
    if not os.path.exists(cache_filename):
        logger.error(f"財務データファイル '{cache_filename}' が見つかりません")
        return pd.DataFrame()
    
    try:
        df = pd.read_csv(
            cache_filename, 
            encoding='utf-8-sig',
            parse_dates=['DisclosedDate'],
            low_memory=False
        )
        
        logger.info(f"財務データ読み込み成功: {len(df):,}件")
        
        # 列名のマッピング
        column_mapping = {
            'IssuedShareTotal': ['IssuedShareTotal', 'NumberOfIssuedAndOutstandingSharesAtTheEndOfFiscalYearIncludingTreasuryStock'],
            'Equity': ['Equity', 'NetAssets', 'TotalEquity'],
            'Profit': ['Profit', 'NetIncome', 'ProfitAttributableToOwnersOfParent']
        }
        
        available_cols = df.columns.tolist()
        required_columns = {}
        
        for target_col, possible_names in column_mapping.items():
            found = False
            for possible_name in possible_names:
                if possible_name in available_cols:
                    required_columns[target_col] = possible_name
                    found = True
                    break
            if not found:
                required_columns[target_col] = None
        
        rename_dict = {v: k for k, v in required_columns.items() if v is not None}
        df = df.rename(columns=rename_dict)
        
        for col in ['IssuedShareTotal', 'Equity', 'Profit']:
            if col not in df.columns:
                df[col] = 0
        
        base_cols = ['Code', 'DisclosedDate']
        if 'CompanyName' in df.columns:
            base_cols.append('CompanyName')
        
        final_cols = base_cols + ['Profit', 'Equity', 'IssuedShareTotal']
        df = df[final_cols].copy()
        
        logger.info(f"使用列: {df.columns.tolist()}")
        
        return df
    
    except Exception as e:
        logger.error(f"財務データ読み込みエラー: {e}", exc_info=True)
        return pd.DataFrame()

# ===================================
# B. 株価データ読み込み（メモリ効率化版・修正）
# ===================================

def load_existing_price_data(ohlcv_dir: str = './OHLCV_Adjusted') -> pd.DataFrame:
    """株価データ読み込み（メモリ効率化版・修正）"""
    global PRICE_CACHE
    
    if not os.path.exists(ohlcv_dir):
        logger.error(f"株価データディレクトリ '{ohlcv_dir}' が見つかりません。")
        return pd.DataFrame()
    
    csv_files = sorted([
        f for f in os.listdir(ohlcv_dir) 
        if f.startswith('OHLCV_Adjusted_') and f.endswith('.csv') and f != 'OHLCV_Adjusted_TOPIX.csv'
    ])
    
    if not csv_files:
        logger.error(f"ディレクトリ '{ohlcv_dir}' 内にCSVファイルが見つかりません。")
        return pd.DataFrame()
    
    logger.info(f"株価ファイル数: {len(csv_files)}個")
    
    all_dataframes = []
    usecols = ['Date', 'Ticker', 'AdjustmentClose']
    
    for csv_file in tqdm(csv_files, desc="株価ファイル読み込み中"):
        file_path = os.path.join(ohlcv_dir, csv_file)
        try:
            df = pd.read_csv(
                file_path,
                usecols=usecols,
                parse_dates=['Date'],
                dtype={'Ticker': 'Int64', 'AdjustmentClose': 'float32'}
            )
            
            # メモリ効率化：各ファイルで変換・フィルタリングしてから結合
            df = df.drop_duplicates(subset=['Ticker', 'Date'], keep='first')
            df = df.dropna(subset=['AdjustmentClose'])
            
            # Code列を変換（メモリ効率的に：fillna → int64 → str）
            df['Ticker'] = df['Ticker'].fillna(0).astype('int64').astype('str').str.zfill(4)
            df = df.rename(columns={'Ticker': 'Code', 'AdjustmentClose': 'Close'})
            
            all_dataframes.append(df)
            
        except Exception as e:
            logger.warning(f"ファイル読み込みエラー ({csv_file}): {e}")
            continue
    
    if not all_dataframes:
        return pd.DataFrame()
    
    logger.info("全ファイルを結合中...")
    df_all = pd.concat(all_dataframes, ignore_index=True)
    
    logger.info(f"結合完了: {len(df_all):,}件")
    
    logger.info("重複除去 & ソート中...")
    df_all = df_all.sort_values(['Code', 'Date'])
    df_all = df_all.drop_duplicates(subset=['Code', 'Date'], keep='first')
    
    logger.info(f"重複除去後: {len(df_all):,}件")
    
    return df_all

# ===================================
# C. TOPIX データ読み込み
# ===================================

def load_topix_data(ohlcv_dir: str = './OHLCV_Adjusted') -> pd.DataFrame:
    """TOPIXデータ読み込み"""
    
    topix_file = os.path.join(ohlcv_dir, 'OHLCV_Adjusted_TOPIX.csv')
    
    if not os.path.exists(topix_file):
        logger.warning(f"TOPIXファイルが見つかりません: {topix_file}")
        return pd.DataFrame()
    
    try:
        df = pd.read_csv(
            topix_file,
            parse_dates=['Date'],
            usecols=['Date', 'AdjustmentClose']
        )
        
        df = df.rename(columns={'AdjustmentClose': 'TOPIX'})
        df = df.sort_values('Date').reset_index(drop=True)
        
        # 200日移動平均を計算
        df['TOPIX_SMA200'] = df['TOPIX'].rolling(window=200, min_periods=1).mean()
        
        logger.info(f"TOPIX データ読み込み完了: {len(df)}日分")
        
        return df
    
    except Exception as e:
        logger.error(f"TOPIX データ読み込みエラー: {e}")
        return pd.DataFrame()

# ===================================
# D. 財務指標計算
# ===================================

def safe_code_to_int(code_series: pd.Series) -> pd.Series:
    """証券コードを整数に変換（ゼロ埋め対応）"""
    cleaned = code_series.astype(str).str.replace(r'\D', '', regex=True)
    cleaned = cleaned.replace('', '0')
    return pd.to_numeric(cleaned, errors='coerce').fillna(0).astype('int64')

def calculate_market_metrics_fast_chunked(statements_df: pd.DataFrame, 
                                          prices_df: pd.DataFrame,
                                          chunk_size: int = 200) -> pd.DataFrame:
    """時価総額・PBR・ROE計算（チャンク処理版）"""
    
    logger.info("時価総額・PBR・ROE計算中（チャンク処理版）...")
    
    statements_df = statements_df.copy()
    statements_df['Profit'] = pd.to_numeric(statements_df['Profit'], errors='coerce').fillna(0)
    statements_df['Equity'] = pd.to_numeric(statements_df['Equity'], errors='coerce').fillna(0)
    statements_df['IssuedShareTotal'] = pd.to_numeric(statements_df['IssuedShareTotal'], errors='coerce').fillna(1)
    
    statements_df = statements_df[
        (statements_df['Equity'] > 0) & 
        (statements_df['IssuedShareTotal'] > 0)
    ]
    logger.info(f"有効な財務データ: {len(statements_df):,}件")
    
    # Code変換
    logger.info("Code列を整数型に変換中...")
    statements_df['Code_int'] = safe_code_to_int(statements_df['Code'])
    prices_df['Code_int'] = safe_code_to_int(prices_df['Code'])
    
    statements_df = statements_df[statements_df['Code_int'] > 0]
    prices_df = prices_df[prices_df['Code_int'] > 0]
    
    # ソート
    statements_df = statements_df.sort_values(['Code_int', 'DisclosedDate']).reset_index(drop=True)
    prices_df = prices_df.sort_values(['Code_int', 'Date']).reset_index(drop=True)
    
    # 重複除去
    statements_df = statements_df.drop_duplicates(subset=['Code_int', 'DisclosedDate'], keep='first')
    prices_df = prices_df.drop_duplicates(subset=['Code_int', 'Date'], keep='first')
    
    logger.info(f"ソート・重複除去後: 財務 {len(statements_df):,}件, 株価 {len(prices_df):,}件")
    
    # 銘柄ごとにマージ
    logger.info("銘柄ごとにマージ中...")
    
    statements_groups = list(statements_df.groupby('Code_int'))
    prices_dict = {code: group for code, group in prices_df.groupby('Code_int')}
    
    merged_list = []
    num_chunks = (len(statements_groups) + chunk_size - 1) // chunk_size
    
    for chunk_idx in tqdm(range(num_chunks), desc="マージ処理"):
        start_idx = chunk_idx * chunk_size
        end_idx = min((chunk_idx + 1) * chunk_size, len(statements_groups))
        
        chunk_groups = statements_groups[start_idx:end_idx]
        
        for code, stmt_code in chunk_groups:
            if code not in prices_dict:
                continue
            
            price_code = prices_dict[code]
            
            if len(price_code) == 0:
                continue
            
            stmt_code = stmt_code.sort_values('DisclosedDate').reset_index(drop=True)
            price_code = price_code.sort_values('Date').reset_index(drop=True)
            
            try:
                merged = pd.merge_asof(
                    stmt_code,
                    price_code[['Date', 'Close']],
                    left_on='DisclosedDate',
                    right_on='Date',
                    direction='backward',
                    tolerance=pd.Timedelta(days=10)
                )
                
                if not merged.empty:
                    merged_list.append(merged)
            except Exception:
                continue
    
    if not merged_list:
        logger.error("マージ結果が空です")
        return pd.DataFrame()
    
    df_merged = pd.concat(merged_list, ignore_index=True)
    df_merged = df_merged.dropna(subset=['Close'])
    
    logger.info(f"マージ完了: {len(df_merged):,}件")
    
    # 財務指標計算
    df_merged['MarketCap'] = df_merged['Close'] * df_merged['IssuedShareTotal']
    df_merged['PBR'] = df_merged['MarketCap'] / df_merged['Equity']
    df_merged['ROE'] = (df_merged['Profit'] / df_merged['Equity']) * 100
    
    result_cols = ['Code', 'DisclosedDate', 'Close', 'MarketCap', 'PBR', 'ROE', 'Date']
    
    if 'CompanyName' in df_merged.columns:
        result_cols.insert(1, 'CompanyName')
    
    result_df = df_merged[result_cols].copy()
    result_df = result_df.rename(columns={'Close': 'StockPrice', 'Date': 'PriceDate'})
    
    # 異常値除去
    mask = (
        (result_df['PBR'] > 0) &
        (result_df['PBR'] < 50) &
        (result_df['ROE'] > -100) &
        (result_df['ROE'] < 100) &
        (result_df['MarketCap'] > 1_000_000_000)
    )
    
    result_df = result_df[mask].copy()
    
    logger.info(f"計算完了: {len(result_df):,}件")
    
    return result_df

# ===================================
# E. ポートフォリオ構築
# ===================================

def build_unit_share_portfolio(stock_candidates: pd.DataFrame,
                               target_positions: int = 20,
                               initial_capital: float = 10_000_000) -> dict:
    """100株単位ポートフォリオ構築"""
    
    if len(stock_candidates) == 0:
        return {
            'stocks': [],
            'shares': [],
            'prices': [],
            'amounts': []
        }
    
    # 上位N銘柄を選択
    selected = stock_candidates.head(target_positions).copy()
    
    # 1銘柄あたりの配分資金
    capital_per_stock = initial_capital / len(selected)
    
    stocks = []
    shares_list = []
    prices_list = []
    amounts_list = []
    
    for _, row in selected.iterrows():
        code = row['Code']
        price = row['StockPrice']
        
        # 100株単位で購入可能な株数を計算
        required_amount = price * UNIT_SHARES
        
        if required_amount <= capital_per_stock:
            shares = int(capital_per_stock // required_amount) * UNIT_SHARES
            
            if shares > 0:
                stocks.append(code)
                shares_list.append(shares)
                prices_list.append(price)
                amounts_list.append(shares * price)
    
    return {
        'stocks': stocks,
        'shares': shares_list,
        'prices': prices_list,
        'amounts': amounts_list
    }

# ===================================
# F. 価格データ前処理（高速化・修正）
# ===================================

def preprocess_price_data(prices_df: pd.DataFrame) -> dict:
    """価格データを高速検索用の辞書に変換"""
    
    logger.info("価格データを辞書化中...")
    
    # Code × Date の辞書を作成
    price_dict = {}
    
    for code, group in tqdm(prices_df.groupby('Code'), desc="価格辞書化"):
        group = group.sort_values('Date')
        price_dict[code] = {
            'dates': group['Date'].values,
            'prices': group['Close'].values
        }
    
    logger.info(f"価格辞書化完了: {len(price_dict)}銘柄")
    
    return price_dict

def get_price_fast(code: str, target_date: pd.Timestamp, price_dict: dict, tolerance_days: int = 2) -> float:
    """高速価格取得（辞書ベース・修正版）"""
    
    if code not in price_dict:
        return np.nan
    
    dates = price_dict[code]['dates']
    prices = price_dict[code]['prices']
    
    # 二分探索で最も近い日付を探す
    target_ts = target_date.value  # ナノ秒単位のTimestamp
    
    # tolerance内の価格を探す
    tolerance_ns = tolerance_days * 24 * 60 * 60 * 10**9
    
    # datesをナノ秒単位の整数配列に変換
    dates_ns = dates.astype('datetime64[ns]').astype('int64')
    
    # 最も近い日付のインデックスを探す
    idx = np.searchsorted(dates_ns, target_ts)
    
    # 前後の日付をチェック
    candidates = []
    
    if idx < len(dates):
        diff = abs(dates_ns[idx] - target_ts)
        candidates.append((diff, prices[idx]))
    
    if idx > 0:
        diff = abs(dates_ns[idx-1] - target_ts)
        candidates.append((diff, prices[idx-1]))
    
    if not candidates:
        return np.nan
    
    # 最も近い価格を返す
    min_diff, best_price = min(candidates, key=lambda x: x[0])
    
    if min_diff <= tolerance_ns:
        return best_price
    else:
        return np.nan

# ===================================
# G. ポートフォリオ管理クラス
# ===================================

class Portfolio:
    """ポートフォリオ管理（取引履歴・税金計算付き）"""
    
    def __init__(self, initial_capital: float):
        self.cash = initial_capital
        self.positions = {}  # {code: {'shares': int, 'buy_price': float}}
        self.total_tax_paid = 0.0
        self.total_trading_cost = 0.0
        self.transaction_history = []
    
    def get_position_value(self, code: str, current_price: float) -> float:
        """保有銘柄の時価評価額"""
        if code not in self.positions:
            return 0.0
        return self.positions[code]['shares'] * current_price
    
    def get_total_value(self, price_dict: dict, current_date: pd.Timestamp) -> float:
        """ポートフォリオ全体の時価評価額"""
        total = self.cash
        
        for code, position in self.positions.items():
            current_price = get_price_fast(code, current_date, price_dict)
            if not np.isnan(current_price):
                total += position['shares'] * current_price
        
        return total
    
    def buy(self, code: str, shares: int, price: float, date: pd.Timestamp, slippage: float = SLIPPAGE):
        """株式購入（スリッページ考慮）"""
        
        if shares <= 0:
            return False
        
        # スリッページを考慮した購入価格
        actual_price = price * (1 + slippage)
        cost = shares * actual_price
        
        if cost > self.cash:
            return False
        
        # 購入実行
        self.cash -= cost
        self.total_trading_cost += cost * slippage
        
        if code in self.positions:
            # 既存ポジションに追加（平均取得単価を更新）
            old_shares = self.positions[code]['shares']
            old_price = self.positions[code]['buy_price']
            new_shares = old_shares + shares
            new_price = (old_shares * old_price + shares * actual_price) / new_shares
            
            self.positions[code] = {
                'shares': new_shares,
                'buy_price': new_price
            }
        else:
            self.positions[code] = {
                'shares': shares,
                'buy_price': actual_price
            }
        
        # 取引履歴
        self.transaction_history.append({
            'date': date,
            'action': 'BUY',
            'code': code,
            'shares': shares,
            'price': actual_price,
            'cost': cost
        })
        
        return True
    
    def sell(self, code: str, shares: int, price: float, date: pd.Timestamp, 
             slippage: float = SLIPPAGE, tax_rate: float = TAX_RATE):
        """株式売却（スリッページ・税金考慮）"""
        
        if code not in self.positions or shares <= 0:
            return False
        
        position = self.positions[code]
        
        if shares > position['shares']:
            shares = position['shares']
        
        # スリッページを考慮した売却価格
        actual_price = price * (1 - slippage)
        proceeds = shares * actual_price
        
        # 取得単価
        buy_price = position['buy_price']
        
        # 損益計算
        profit = (actual_price - buy_price) * shares
        tax = 0.0
        
        if profit > 0:
            # 利益が出た場合のみ課税
            tax = profit * tax_rate
            self.total_tax_paid += tax
        
        # 売却実行
        net_proceeds = proceeds - tax
        self.cash += net_proceeds
        self.total_trading_cost += proceeds * slippage
        
        # ポジション更新
        position['shares'] -= shares
        
        if position['shares'] <= 0:
            del self.positions[code]
        
        # 取引履歴
        self.transaction_history.append({
            'date': date,
            'action': 'SELL',
            'code': code,
            'shares': shares,
            'price': actual_price,
            'proceeds': net_proceeds,
            'profit': profit,
            'tax': tax
        })
        
        return True
    
    def get_current_holdings(self) -> list:
        """現在の保有銘柄リスト"""
        return list(self.positions.keys())
    
    def rebalance_to_target(self, target_portfolio: dict, price_dict: dict, 
                           current_date: pd.Timestamp, slippage: float = SLIPPAGE):
        """ターゲットポートフォリオへリバランス（差分取引）"""
        
        target_stocks = set(target_portfolio['stocks'])
        current_stocks = set(self.get_current_holdings())
        
        # 売却すべき銘柄
        to_sell = current_stocks - target_stocks
        
        for code in to_sell:
            current_price = get_price_fast(code, current_date, price_dict)
            if not np.isnan(current_price):
                shares = self.positions[code]['shares']
                self.sell(code, shares, current_price, current_date, slippage)
        
        # 購入すべき銘柄
        target_dict = {
            code: {'shares': shares, 'price': price}
            for code, shares, price in zip(
                target_portfolio['stocks'],
                target_portfolio['shares'],
                target_portfolio['prices']
            )
        }
        
        for code in target_stocks:
            target_shares = target_dict[code]['shares']
            target_price = target_dict[code]['price']
            
            current_shares = self.positions.get(code, {}).get('shares', 0)
            
            if target_shares > current_shares:
                # 買い増し
                to_buy = target_shares - current_shares
                self.buy(code, to_buy, target_price, current_date, slippage)

# ===================================
# H. 日次バックテスト（完全版）
# ===================================

def smbc_value_quality_backtest_daily_complete(
    enhanced_financial_data: pd.DataFrame, 
    prices_df: pd.DataFrame,
    topix_df: pd.DataFrame,
    initial_capital: float = 10_000_000,
    rebalance_frequency: int = 1,
    slippage: float = SLIPPAGE,
    tax_rate: float = TAX_RATE,
    use_market_filter: bool = True
) -> tuple:
    """
    SMBC割安高質戦略（完全版：税金・スリッページ・市況フィルター対応）
    
    Returns:
        (results_df, portfolio, transactions_df)
    """
    
    logger.info(f"日次リバランス戦略バックテスト実行中（完全版）...")
    logger.info(f"  スリッページ: {slippage*100:.2f}%")
    logger.info(f"  税率: {tax_rate*100:.3f}%")
    logger.info(f"  市況フィルター: {'有効' if use_market_filter else '無効'}")
    
    # 価格データを辞書化（事前処理）
    price_dict = preprocess_price_data(prices_df)
    
    # 全営業日を取得
    all_dates = pd.to_datetime(prices_df['Date'].unique())
    all_dates = pd.DatetimeIndex(sorted(all_dates))
    
    # バックテスト期間
    start_date = pd.Timestamp('2016-10-01')
    end_date = pd.Timestamp('2025-10-01')
    
    trading_dates = all_dates[(all_dates >= start_date) & (all_dates <= end_date)]
    
    logger.info(f"バックテスト期間: {start_date.date()} 〜 {end_date.date()}")
    logger.info(f"営業日数: {len(trading_dates)}日")
    logger.info(f"リバランス頻度: {rebalance_frequency}営業日ごと")
    
    # 財務データのキャッシュ
    financial_cache = enhanced_financial_data.sort_values('DisclosedDate').copy()
    
    # TOPIX辞書化
    topix_dict = {}
    if not topix_df.empty:
        for _, row in topix_df.iterrows():
            topix_dict[row['Date']] = {
                'topix': row['TOPIX'],
                'sma200': row['TOPIX_SMA200']
            }
    
    # ポートフォリオ初期化
    portfolio = Portfolio(initial_capital)
    
    last_rebalance_date = None
    rebalance_counter = 0
    
    strategy_results = []
    
    # 日次ループ
    for date_idx, current_date in enumerate(tqdm(trading_dates, desc="日次バックテスト（完全版）")):
        
        # 市況フィルター
        target_positions = 20  # デフォルト
        
        if use_market_filter and current_date in topix_dict:
            topix_info = topix_dict[current_date]
            
            if topix_info['topix'] < topix_info['sma200']:
                # 市況悪化 → ポジション削減
                target_positions = 10
                logger.debug(f"{current_date.date()}: 市況フィルター発動（TOPIX < SMA200）")
        
        # リバランス判定
        should_rebalance = False
        
        if last_rebalance_date is None:
            should_rebalance = True
        elif rebalance_counter >= rebalance_frequency:
            should_rebalance = True
        
        # リバランス実行
        if should_rebalance:
            current_data = financial_cache[financial_cache['DisclosedDate'] <= current_date]
            
            if len(current_data) > 0:
                current_data = current_data.groupby('Code', as_index=False).last()
            
            if len(current_data) >= 100:
                # PBRとROEでランキング
                current_data['PBR_Rank'] = current_data['PBR'].rank(method='first', ascending=True)
                current_data['ROE_Rank'] = current_data['ROE'].rank(method='first', ascending=False)
                
                # 四分位分割
                current_data['PBR_Quartile'] = pd.qcut(current_data['PBR_Rank'], q=4, labels=[1, 2, 3, 4], duplicates='drop')
                current_data['ROE_Quartile'] = pd.qcut(current_data['ROE_Rank'], q=4, labels=[1, 2, 3, 4], duplicates='drop')
                
                # ロング候補
                long_candidates = current_data[
                    (current_data['PBR_Quartile'] == 1) & 
                    (current_data['ROE_Quartile'] == 4)
                ].nsmallest(50, 'PBR')
                
                # 現在の総資産で再計算
                current_capital = portfolio.get_total_value(price_dict, current_date)
                
                # ターゲットポートフォリオ構築
                target_portfolio = build_unit_share_portfolio(
                    long_candidates,
                    target_positions=target_positions,
                    initial_capital=current_capital
                )
                
                # リバランス実行（差分取引）
                portfolio.rebalance_to_target(
                    target_portfolio, 
                    price_dict, 
                    current_date, 
                    slippage
                )
                
                last_rebalance_date = current_date
                rebalance_counter = 0
        
        # 日次パフォーマンス記録
        total_value = portfolio.get_total_value(price_dict, current_date)
        
        daily_return = 0.0
        if date_idx > 0 and len(strategy_results) > 0:
            prev_value = strategy_results[-1]['total_value']
            if prev_value > 0:
                daily_return = (total_value - prev_value) / prev_value
        
        # 結果記録
        strategy_results.append({
            'date': current_date,
            'total_value': total_value,
            'cash': portfolio.cash,
            'daily_return': daily_return,
            'num_positions': len(portfolio.positions),
            'cumulative_tax': portfolio.total_tax_paid,
            'cumulative_trading_cost': portfolio.total_trading_cost,
            'is_rebalance_day': should_rebalance,
            'target_positions': target_positions
        })
        
        rebalance_counter += 1
    
    results_df = pd.DataFrame(strategy_results)
    transactions_df = pd.DataFrame(portfolio.transaction_history)
    
    return results_df, portfolio, transactions_df

# ===================================
# I. パフォーマンス分析
# ===================================

def calculate_annual_performance(returns: pd.Series, dates: pd.Series = None) -> float:
    """年率リターン計算"""
    if len(returns) == 0:
        return 0.0
    
    cumulative = (1 + returns).prod()
    
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
    else:
        years = len(returns) / 252  # 営業日ベース
    
    if years > 0:
        return ((cumulative ** (1/years)) - 1) * 100
    else:
        return 0.0

def calculate_volatility_annualized(returns: pd.Series, dates: pd.Series = None) -> float:
    """年率換算ボラティリティ計算"""
    if len(returns) == 0:
        return 0.0
    
    period_std = returns.std()
    
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
        periods_per_year = len(returns) / years if years > 0 else 252
    else:
        periods_per_year = 252  # 営業日ベース
    
    if periods_per_year > 0:
        return period_std * np.sqrt(periods_per_year) * 100
    else:
        return 0.0

def calculate_sharpe_ratio(returns: pd.Series, risk_free_rate: float = 0.0, dates: pd.Series = None) -> float:
    """シャープレシオ計算（年率換算）"""
    if len(returns) == 0 or returns.std() == 0:
        return 0.0
    
    excess_returns = returns - risk_free_rate / 252  # 日次無リスク金利
    mean_return = excess_returns.mean()
    std_return = excess_returns.std()
    
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
        periods_per_year = len(returns) / years if years > 0 else 252
    else:
        periods_per_year = 252
    
    if std_return > 0:
        return (mean_return / std_return) * np.sqrt(periods_per_year)
    else:
        return 0.0

def calculate_max_drawdown(returns: pd.Series, dates: pd.Series = None) -> dict:
    """最大ドローダウン計算"""
    if len(returns) == 0:
        return {
            'max_drawdown': 0.0,
            'mdd_date': None,
            'mdd_period': None
        }
    
    cumulative = (1 + returns).cumprod()
    running_max = cumulative.cummax()
    drawdown = (cumulative - running_max) / running_max
    
    min_dd_idx = drawdown.idxmin()
    max_dd_value = drawdown.min() * 100
    
    mdd_date = None
    if dates is not None and len(dates) > 0:
        if min_dd_idx in dates.index:
            mdd_date = dates.loc[min_dd_idx]
        elif isinstance(min_dd_idx, (int, np.integer)) and 0 <= min_dd_idx < len(dates):
            mdd_date = dates.iloc[min_dd_idx]
    
    return {
        'max_drawdown': max_dd_value,
        'mdd_date': mdd_date,
        'mdd_period': None
    }

def calculate_additional_metrics_complete(results_df: pd.DataFrame, portfolio: Portfolio) -> dict:
    """追加パフォーマンス指標の計算（完全版）"""
    
    returns = results_df['daily_return']
    dates = results_df.get('date', None)
    
    # MDD計算
    mdd_result = calculate_max_drawdown(returns, dates)
    
    initial_capital = results_df['total_value'].iloc[0]
    final_capital = results_df['total_value'].iloc[-1]
    
    metrics = {
        'initial_capital': initial_capital,
        'final_capital': final_capital,
        'total_return': (final_capital / initial_capital - 1) * 100,
        'annual_return': calculate_annual_performance(returns, dates),
        'sharpe_ratio': calculate_sharpe_ratio(returns, risk_free_rate=0.0, dates=dates),
        'max_drawdown': mdd_result['max_drawdown'],
        'mdd_date': mdd_result['mdd_date'],
        'win_rate': (returns > 0).sum() / len(returns) * 100,
        'avg_win': returns[returns > 0].mean() * 100 if (returns > 0).any() else 0,
        'avg_loss': returns[returns < 0].mean() * 100 if (returns < 0).any() else 0,
        'volatility': calculate_volatility_annualized(returns, dates),
        'total_tax_paid': portfolio.total_tax_paid,
        'total_trading_cost': portfolio.total_trading_cost,
        'tax_rate_on_return': (portfolio.total_tax_paid / (final_capital - initial_capital)) * 100 if final_capital > initial_capital else 0,
        'num_transactions': len(portfolio.transaction_history),
    }
    
    return metrics

# ===================================
# J. 結果出力
# ===================================

def print_performance_report_complete(results_df: pd.DataFrame, metrics: dict, portfolio: Portfolio):
    """パフォーマンスレポート出力（完全版）"""
    
    print("\n" + "=" * 80)
    print("📊 日次リバランス戦略パフォーマンス結果（完全版）")
    print("=" * 80)
    
    print(f"\n【戦略設定】")
    print(f"  ポジション:          ロングのみ（ショートなし）")
    print(f"  リバランス頻度:      日次")
    print(f"  制約:                100株単位購入制限")
    print(f"  スリッページ:        0.3%")
    print(f"  税率:                20.315%")
    print(f"  市況フィルター:      有効（TOPIX < SMA200 で減少）")
    print(f"  平均ポジション数:    {results_df['num_positions'].mean():.1f}銘柄")
    
    print(f"\n【パフォーマンス】")
    print(f"  初期資本:            ¥{metrics['initial_capital']:,.0f}")
    print(f"  最終資本:            ¥{metrics['final_capital']:,.0f}")
    print(f"  累積リターン:        {metrics['total_return']:.2f}%")
    print(f"  年率リターン:        {metrics['annual_return']:.2f}%")
    print(f"  シャープレシオ:      {metrics['sharpe_ratio']:.2f}")
    print(f"  最大DD:              {metrics['max_drawdown']:.2f}%")
    if metrics.get('mdd_date'):
        mdd_date_str = metrics['mdd_date'].strftime('%Y年%m月%d日')
        print(f"  MDD発生日:          {mdd_date_str}")
    print(f"  年率ボラティリティ:  {metrics['volatility']:.2f}%")
    
    print(f"\n【コスト分析】")
    print(f"  総取引回数:          {metrics['num_transactions']:,}回")
    print(f"  累積税金:            ¥{metrics['total_tax_paid']:,.0f}")
    print(f"  累積取引コスト:      ¥{metrics['total_trading_cost']:,.0f}")
    print(f"  税金率（対利益）:    {metrics['tax_rate_on_return']:.2f}%")
    print(f"  総コスト:            ¥{metrics['total_tax_paid'] + metrics['total_trading_cost']:,.0f}")
    
    print(f"\n【勝率・損益（日次）】")
    print(f"  勝率:                {metrics['win_rate']:.1f}%")
    print(f"  平均利益:            {metrics['avg_win']:.3f}%")
    print(f"  平均損失:            {metrics['avg_loss']:.3f}%")
    
    print(f"\n【比較】")
    print(f"  年次版年率リターン:  28.52%")
    print(f"  日次版年率リターン:  {metrics['annual_return']:.2f}%")
    print(f"  差分:                {metrics['annual_return'] - 28.52:+.2f}%")
    
    print("\n" + "=" * 80)

# ===================================
# K. メイン実行
# ===================================

if __name__ == "__main__":
    
    print("=" * 80)
    print("🚀 SMBC「割安高質」戦略 - 日次リバランス版（完全版）")
    print("=" * 80)
    print(f"\n📋 検証概要:")
    print(f"  戦略名: 割安高質（ロングオンリー）")
    print(f"  検証期間: 2016年10月〜2025年10月")
    print(f"  リバランス: 日次")
    print(f"  スリッページ: 0.3%")
    print(f"  税率: 20.315%")
    print(f"  市況フィルター: 有効（TOPIX基準）")
    print(f"  実行日時: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    print("=" * 80)
    
    # データ読み込み
    statements_df = load_financial_data()
    if statements_df.empty:
        logger.error("財務データ読み込み失敗")
        exit(1)
    
    prices_df = load_existing_price_data()
    if prices_df.empty:
        logger.error("株価データ読み込み失敗")
        exit(1)
    
    # TOPIX データ読み込み
    topix_df = load_topix_data()
    
    # 財務指標計算
    enhanced_financial_data = calculate_market_metrics_fast_chunked(statements_df, prices_df, chunk_size=200)
    if enhanced_financial_data.empty:
        logger.error("財務指標計算失敗")
        exit(1)
    
    # 日次バックテスト実行（完全版）
    results, portfolio, transactions = smbc_value_quality_backtest_daily_complete(
        enhanced_financial_data, 
        prices_df,
        topix_df,
        initial_capital=10_000_000,
        rebalance_frequency=1,
        slippage=0.003,
        tax_rate=0.20315,
        use_market_filter=True
    )
    
    if results.empty:
        logger.error("バックテスト実行失敗")
        exit(1)
    
    # パフォーマンス指標計算
    metrics = calculate_additional_metrics_complete(results, portfolio)
    
    # 結果出力
    print_performance_report_complete(results, metrics, portfolio)
    
    # 結果保存
    results.to_csv('backtest_results_daily_complete.csv', index=False, encoding='utf-8-sig')
    transactions.to_csv('transactions_daily_complete.csv', index=False, encoding='utf-8-sig')
    
    logger.info("結果をCSVに保存: backtest_results_daily_complete.csv")
    logger.info("取引履歴を保存: transactions_daily_complete.csv")
    
    print("\n" + "=" * 80)
    print("🏆 日次リバランス戦略検証完了（完全版）")
    print("=" * 80)
    print(f"\n📁 出力ファイル:")
    print(f"  - バックテスト結果: backtest_results_daily_complete.csv")
    print(f"  - 取引履歴: transactions_daily_complete.csv")
    print(f"  - ログファイル: backtest_daily_value_quality_complete.log")
    print("\n")
```

```python
import pandas as pd
import numpy as np
from tqdm import tqdm 
import warnings
import os
import logging
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import datetime

warnings.filterwarnings('ignore')

# ===================================
# ロギング設定
# ===================================
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[
        logging.FileHandler('backtest_daily_value_quality_complete.log', encoding='utf-8'),
        logging.StreamHandler()
    ]
)
logger = logging.getLogger(__name__)

# ===================================
# グローバル変数
# ===================================
CACHE_FILE = 'topix_quarterly_statements.csv'
PRICE_CACHE = {}

# 税金パラメータ
TAX_RATE = 0.20315  # 譲渡益税 20.315%

# 単位株制限
UNIT_SHARES = 100  # 100株単位

# スリッページ
SLIPPAGE = 0.003  # 0.3%

# ===================================
# A. 財務データ読み込み
# ===================================

def load_financial_data(cache_filename: str = CACHE_FILE) -> pd.DataFrame:
    """財務諸表データをキャッシュから読み込み（列名を自動判定）"""
    
    if not os.path.exists(cache_filename):
        logger.error(f"財務データファイル '{cache_filename}' が見つかりません")
        return pd.DataFrame()
    
    try:
        df = pd.read_csv(
            cache_filename, 
            encoding='utf-8-sig',
            parse_dates=['DisclosedDate'],
            low_memory=False
        )
        
        logger.info(f"財務データ読み込み成功: {len(df):,}件")
        
        # 列名のマッピング
        column_mapping = {
            'IssuedShareTotal': ['IssuedShareTotal', 'NumberOfIssuedAndOutstandingSharesAtTheEndOfFiscalYearIncludingTreasuryStock'],
            'Equity': ['Equity', 'NetAssets', 'TotalEquity'],
            'Profit': ['Profit', 'NetIncome', 'ProfitAttributableToOwnersOfParent']
        }
        
        available_cols = df.columns.tolist()
        required_columns = {}
        
        for target_col, possible_names in column_mapping.items():
            found = False
            for possible_name in possible_names:
                if possible_name in available_cols:
                    required_columns[target_col] = possible_name
                    found = True
                    break
            if not found:
                required_columns[target_col] = None
        
        rename_dict = {v: k for k, v in required_columns.items() if v is not None}
        df = df.rename(columns=rename_dict)
        
        for col in ['IssuedShareTotal', 'Equity', 'Profit']:
            if col not in df.columns:
                df[col] = 0
        
        base_cols = ['Code', 'DisclosedDate']
        if 'CompanyName' in df.columns:
            base_cols.append('CompanyName')
        
        final_cols = base_cols + ['Profit', 'Equity', 'IssuedShareTotal']
        df = df[final_cols].copy()
        
        logger.info(f"使用列: {df.columns.tolist()}")
        
        return df
    
    except Exception as e:
        logger.error(f"財務データ読み込みエラー: {e}", exc_info=True)
        return pd.DataFrame()

# ===================================
# B. 株価データ読み込み
# ===================================

def load_existing_price_data(ohlcv_dir: str = './OHLCV_Adjusted') -> pd.DataFrame:
    """株価データ読み込み（メモリ効率化版）"""
    global PRICE_CACHE
    
    if not os.path.exists(ohlcv_dir):
        logger.error(f"株価データディレクトリ '{ohlcv_dir}' が見つかりません。")
        return pd.DataFrame()
    
    csv_files = sorted([
        f for f in os.listdir(ohlcv_dir) 
        if f.startswith('OHLCV_Adjusted_') and f.endswith('.csv') and f != 'OHLCV_Adjusted_TOPIX.csv'
    ])
    
    if not csv_files:
        logger.error(f"ディレクトリ '{ohlcv_dir}' 内にCSVファイルが見つかりません。")
        return pd.DataFrame()
    
    logger.info(f"株価ファイル数: {len(csv_files)}個")
    
    all_dataframes = []
    usecols = ['Date', 'Ticker', 'AdjustmentClose']
    
    for csv_file in tqdm(csv_files, desc="株価ファイル読み込み中"):
        file_path = os.path.join(ohlcv_dir, csv_file)
        try:
            df = pd.read_csv(
                file_path,
                usecols=usecols,
                parse_dates=['Date'],
                dtype={'Ticker': 'Int64', 'AdjustmentClose': 'float32'}
            )
            
            df = df.drop_duplicates(subset=['Ticker', 'Date'], keep='first')
            all_dataframes.append(df)
            
        except Exception as e:
            logger.warning(f"ファイル読み込みエラー ({csv_file}): {e}")
            continue
    
    if not all_dataframes:
        return pd.DataFrame()
    
    logger.info("全ファイルを結合中...")
    df_all = pd.concat(all_dataframes, ignore_index=True)
    
    logger.info(f"結合完了: {len(df_all):,}件")
    
    df_all = df_all.rename(columns={'Ticker': 'Code', 'AdjustmentClose': 'Close'})
    df_all['Code'] = df_all['Code'].astype('str').str.replace('<NA>', '0').str.zfill(4)
    df_all = df_all.dropna(subset=['Close'])
    
    logger.info("重複除去 & ソート中...")
    df_all = df_all.sort_values(['Code', 'Date'])
    df_all = df_all.drop_duplicates(subset=['Code', 'Date'], keep='first')
    
    logger.info(f"重複除去後: {len(df_all):,}件")
    
    return df_all

# ===================================
# C. TOPIX データ読み込み
# ===================================

def load_topix_data(ohlcv_dir: str = './OHLCV_Adjusted') -> pd.DataFrame:
    """TOPIXデータ読み込み"""
    
    topix_file = os.path.join(ohlcv_dir, 'OHLCV_Adjusted_TOPIX.csv')
    
    if not os.path.exists(topix_file):
        logger.warning(f"TOPIXファイルが見つかりません: {topix_file}")
        return pd.DataFrame()
    
    try:
        df = pd.read_csv(
            topix_file,
            parse_dates=['Date'],
            usecols=['Date', 'AdjustmentClose']
        )
        
        df = df.rename(columns={'AdjustmentClose': 'TOPIX'})
        df = df.sort_values('Date').reset_index(drop=True)
        
        # 200日移動平均を計算
        df['TOPIX_SMA200'] = df['TOPIX'].rolling(window=200, min_periods=1).mean()
        
        logger.info(f"TOPIX データ読み込み完了: {len(df)}日分")
        
        return df
    
    except Exception as e:
        logger.error(f"TOPIX データ読み込みエラー: {e}")
        return pd.DataFrame()

# ===================================
# D. 財務指標計算
# ===================================

def safe_code_to_int(code_series: pd.Series) -> pd.Series:
    """証券コードを整数に変換（ゼロ埋め対応）"""
    cleaned = code_series.astype(str).str.replace(r'\D', '', regex=True)
    cleaned = cleaned.replace('', '0')
    return pd.to_numeric(cleaned, errors='coerce').fillna(0).astype('int64')

def calculate_market_metrics_fast_chunked(statements_df: pd.DataFrame, 
                                          prices_df: pd.DataFrame,
                                          chunk_size: int = 200) -> pd.DataFrame:
    """時価総額・PBR・ROE計算（チャンク処理版）"""
    
    logger.info("時価総額・PBR・ROE計算中（チャンク処理版）...")
    
    statements_df = statements_df.copy()
    statements_df['Profit'] = pd.to_numeric(statements_df['Profit'], errors='coerce').fillna(0)
    statements_df['Equity'] = pd.to_numeric(statements_df['Equity'], errors='coerce').fillna(0)
    statements_df['IssuedShareTotal'] = pd.to_numeric(statements_df['IssuedShareTotal'], errors='coerce').fillna(1)
    
    statements_df = statements_df[
        (statements_df['Equity'] > 0) & 
        (statements_df['IssuedShareTotal'] > 0)
    ]
    logger.info(f"有効な財務データ: {len(statements_df):,}件")
    
    # Code変換
    logger.info("Code列を整数型に変換中...")
    statements_df['Code_int'] = safe_code_to_int(statements_df['Code'])
    prices_df['Code_int'] = safe_code_to_int(prices_df['Code'])
    
    statements_df = statements_df[statements_df['Code_int'] > 0]
    prices_df = prices_df[prices_df['Code_int'] > 0]
    
    # ソート
    statements_df = statements_df.sort_values(['Code_int', 'DisclosedDate']).reset_index(drop=True)
    prices_df = prices_df.sort_values(['Code_int', 'Date']).reset_index(drop=True)
    
    # 重複除去
    statements_df = statements_df.drop_duplicates(subset=['Code_int', 'DisclosedDate'], keep='first')
    prices_df = prices_df.drop_duplicates(subset=['Code_int', 'Date'], keep='first')
    
    logger.info(f"ソート・重複除去後: 財務 {len(statements_df):,}件, 株価 {len(prices_df):,}件")
    
    # 銘柄ごとにマージ
    logger.info("銘柄ごとにマージ中...")
    
    statements_groups = list(statements_df.groupby('Code_int'))
    prices_dict = {code: group for code, group in prices_df.groupby('Code_int')}
    
    merged_list = []
    num_chunks = (len(statements_groups) + chunk_size - 1) // chunk_size
    
    for chunk_idx in tqdm(range(num_chunks), desc="マージ処理"):
        start_idx = chunk_idx * chunk_size
        end_idx = min((chunk_idx + 1) * chunk_size, len(statements_groups))
        
        chunk_groups = statements_groups[start_idx:end_idx]
        
        for code, stmt_code in chunk_groups:
            if code not in prices_dict:
                continue
            
            price_code = prices_dict[code]
            
            if len(price_code) == 0:
                continue
            
            stmt_code = stmt_code.sort_values('DisclosedDate').reset_index(drop=True)
            price_code = price_code.sort_values('Date').reset_index(drop=True)
            
            try:
                merged = pd.merge_asof(
                    stmt_code,
                    price_code[['Date', 'Close']],
                    left_on='DisclosedDate',
                    right_on='Date',
                    direction='backward',
                    tolerance=pd.Timedelta(days=10)
                )
                
                if not merged.empty:
                    merged_list.append(merged)
            except Exception:
                continue
    
    if not merged_list:
        logger.error("マージ結果が空です")
        return pd.DataFrame()
    
    df_merged = pd.concat(merged_list, ignore_index=True)
    df_merged = df_merged.dropna(subset=['Close'])
    
    logger.info(f"マージ完了: {len(df_merged):,}件")
    
    # 財務指標計算
    df_merged['MarketCap'] = df_merged['Close'] * df_merged['IssuedShareTotal']
    df_merged['PBR'] = df_merged['MarketCap'] / df_merged['Equity']
    df_merged['ROE'] = (df_merged['Profit'] / df_merged['Equity']) * 100
    
    result_cols = ['Code', 'DisclosedDate', 'Close', 'MarketCap', 'PBR', 'ROE', 'Date']
    
    if 'CompanyName' in df_merged.columns:
        result_cols.insert(1, 'CompanyName')
    
    result_df = df_merged[result_cols].copy()
    result_df = result_df.rename(columns={'Close': 'StockPrice', 'Date': 'PriceDate'})
    
    # 異常値除去
    mask = (
        (result_df['PBR'] > 0) &
        (result_df['PBR'] < 50) &
        (result_df['ROE'] > -100) &
        (result_df['ROE'] < 100) &
        (result_df['MarketCap'] > 1_000_000_000)
    )
    
    result_df = result_df[mask].copy()
    
    logger.info(f"計算完了: {len(result_df):,}件")
    
    return result_df

# ===================================
# E. ポートフォリオ構築
# ===================================

def build_unit_share_portfolio(stock_candidates: pd.DataFrame,
                               target_positions: int = 20,
                               initial_capital: float = 10_000_000) -> dict:
    """100株単位ポートフォリオ構築"""
    
    if len(stock_candidates) == 0:
        return {
            'stocks': [],
            'shares': [],
            'prices': [],
            'amounts': []
        }
    
    # 上位N銘柄を選択
    selected = stock_candidates.head(target_positions).copy()
    
    # 1銘柄あたりの配分資金
    capital_per_stock = initial_capital / len(selected)
    
    stocks = []
    shares_list = []
    prices_list = []
    amounts_list = []
    
    for _, row in selected.iterrows():
        code = row['Code']
        price = row['StockPrice']
        
        # 100株単位で購入可能な株数を計算
        required_amount = price * UNIT_SHARES
        
        if required_amount <= capital_per_stock:
            shares = int(capital_per_stock // required_amount) * UNIT_SHARES
            
            if shares > 0:
                stocks.append(code)
                shares_list.append(shares)
                prices_list.append(price)
                amounts_list.append(shares * price)
    
    return {
        'stocks': stocks,
        'shares': shares_list,
        'prices': prices_list,
        'amounts': amounts_list
    }

# ===================================
# F. 価格データ前処理（高速化）
# ===================================

def preprocess_price_data(prices_df: pd.DataFrame) -> dict:
    """価格データを高速検索用の辞書に変換"""
    
    logger.info("価格データを辞書化中...")
    
    # Code × Date の辞書を作成
    price_dict = {}
    
    for code, group in tqdm(prices_df.groupby('Code'), desc="価格辞書化"):
        group = group.sort_values('Date')
        price_dict[code] = {
            'dates': group['Date'].values,
            'prices': group['Close'].values
        }
    
    logger.info(f"価格辞書化完了: {len(price_dict)}銘柄")
    
    return price_dict

def get_price_fast(code: str, target_date: pd.Timestamp, price_dict: dict, tolerance_days: int = 2) -> float:
    """高速価格取得（辞書ベース）"""
    
    if code not in price_dict:
        return np.nan
    
    dates = price_dict[code]['dates']
    prices = price_dict[code]['prices']
    
    # 二分探索で最も近い日付を探す
    target_ts = target_date.value  # ナノ秒単位のTimestamp
    
    # tolerance内の価格を探す
    tolerance_ns = tolerance_days * 24 * 60 * 60 * 10**9
    
    # 最も近い日付のインデックスを探す
    idx = np.searchsorted(dates, target_ts)
    
    # 前後の日付をチェック
    candidates = []
    
    if idx < len(dates):
        candidates.append((abs(dates[idx] - target_ts), prices[idx]))
    
    if idx > 0:
        candidates.append((abs(dates[idx-1] - target_ts), prices[idx-1]))
    
    if not candidates:
        return np.nan
    
    # 最も近い価格を返す
    min_diff, best_price = min(candidates, key=lambda x: x[0])
    
    if min_diff <= tolerance_ns:
        return best_price
    else:
        return np.nan

# ===================================
# G. ポートフォリオ管理クラス
# ===================================

class Portfolio:
    """ポートフォリオ管理（取引履歴・税金計算付き）"""
    
    def __init__(self, initial_capital: float):
        self.cash = initial_capital
        self.positions = {}  # {code: {'shares': int, 'buy_price': float}}
        self.total_tax_paid = 0.0
        self.total_trading_cost = 0.0
        self.transaction_history = []
    
    def get_position_value(self, code: str, current_price: float) -> float:
        """保有銘柄の時価評価額"""
        if code not in self.positions:
            return 0.0
        return self.positions[code]['shares'] * current_price
    
    def get_total_value(self, price_dict: dict, current_date: pd.Timestamp) -> float:
        """ポートフォリオ全体の時価評価額"""
        total = self.cash
        
        for code, position in self.positions.items():
            current_price = get_price_fast(code, current_date, price_dict)
            if not np.isnan(current_price):
                total += position['shares'] * current_price
        
        return total
    
    def buy(self, code: str, shares: int, price: float, date: pd.Timestamp, slippage: float = SLIPPAGE):
        """株式購入（スリッページ考慮）"""
        
        if shares <= 0:
            return False
        
        # スリッページを考慮した購入価格
        actual_price = price * (1 + slippage)
        cost = shares * actual_price
        
        if cost > self.cash:
            return False
        
        # 購入実行
        self.cash -= cost
        self.total_trading_cost += cost * slippage
        
        if code in self.positions:
            # 既存ポジションに追加（平均取得単価を更新）
            old_shares = self.positions[code]['shares']
            old_price = self.positions[code]['buy_price']
            new_shares = old_shares + shares
            new_price = (old_shares * old_price + shares * actual_price) / new_shares
            
            self.positions[code] = {
                'shares': new_shares,
                'buy_price': new_price
            }
        else:
            self.positions[code] = {
                'shares': shares,
                'buy_price': actual_price
            }
        
        # 取引履歴
        self.transaction_history.append({
            'date': date,
            'action': 'BUY',
            'code': code,
            'shares': shares,
            'price': actual_price,
            'cost': cost
        })
        
        return True
    
    def sell(self, code: str, shares: int, price: float, date: pd.Timestamp, 
             slippage: float = SLIPPAGE, tax_rate: float = TAX_RATE):
        """株式売却（スリッページ・税金考慮）"""
        
        if code not in self.positions or shares <= 0:
            return False
        
        position = self.positions[code]
        
        if shares > position['shares']:
            shares = position['shares']
        
        # スリッページを考慮した売却価格
        actual_price = price * (1 - slippage)
        proceeds = shares * actual_price
        
        # 取得単価
        buy_price = position['buy_price']
        
        # 損益計算
        profit = (actual_price - buy_price) * shares
        tax = 0.0
        
        if profit > 0:
            # 利益が出た場合のみ課税
            tax = profit * tax_rate
            self.total_tax_paid += tax
        
        # 売却実行
        net_proceeds = proceeds - tax
        self.cash += net_proceeds
        self.total_trading_cost += proceeds * slippage
        
        # ポジション更新
        position['shares'] -= shares
        
        if position['shares'] <= 0:
            del self.positions[code]
        
        # 取引履歴
        self.transaction_history.append({
            'date': date,
            'action': 'SELL',
            'code': code,
            'shares': shares,
            'price': actual_price,
            'proceeds': net_proceeds,
            'profit': profit,
            'tax': tax
        })
        
        return True
    
    def get_current_holdings(self) -> list:
        """現在の保有銘柄リスト"""
        return list(self.positions.keys())
    
    def rebalance_to_target(self, target_portfolio: dict, price_dict: dict, 
                           current_date: pd.Timestamp, slippage: float = SLIPPAGE):
        """ターゲットポートフォリオへリバランス（差分取引）"""
        
        target_stocks = set(target_portfolio['stocks'])
        current_stocks = set(self.get_current_holdings())
        
        # 売却すべき銘柄
        to_sell = current_stocks - target_stocks
        
        for code in to_sell:
            current_price = get_price_fast(code, current_date, price_dict)
            if not np.isnan(current_price):
                shares = self.positions[code]['shares']
                self.sell(code, shares, current_price, current_date, slippage)
        
        # 購入すべき銘柄
        target_dict = {
            code: {'shares': shares, 'price': price}
            for code, shares, price in zip(
                target_portfolio['stocks'],
                target_portfolio['shares'],
                target_portfolio['prices']
            )
        }
        
        for code in target_stocks:
            target_shares = target_dict[code]['shares']
            target_price = target_dict[code]['price']
            
            current_shares = self.positions.get(code, {}).get('shares', 0)
            
            if target_shares > current_shares:
                # 買い増し
                to_buy = target_shares - current_shares
                self.buy(code, to_buy, target_price, current_date, slippage)

# ===================================
# H. 日次バックテスト（完全版）
# ===================================

def smbc_value_quality_backtest_daily_complete(
    enhanced_financial_data: pd.DataFrame, 
    prices_df: pd.DataFrame,
    topix_df: pd.DataFrame,
    initial_capital: float = 10_000_000,
    rebalance_frequency: int = 1,
    slippage: float = SLIPPAGE,
    tax_rate: float = TAX_RATE,
    use_market_filter: bool = True
) -> tuple:
    """
    SMBC割安高質戦略（完全版：税金・スリッページ・市況フィルター対応）
    
    Returns:
        (results_df, portfolio, transactions_df)
    """
    
    logger.info(f"日次リバランス戦略バックテスト実行中（完全版）...")
    logger.info(f"  スリッページ: {slippage*100:.2f}%")
    logger.info(f"  税率: {tax_rate*100:.3f}%")
    logger.info(f"  市況フィルター: {'有効' if use_market_filter else '無効'}")
    
    # 価格データを辞書化（事前処理）
    price_dict = preprocess_price_data(prices_df)
    
    # 全営業日を取得
    all_dates = pd.to_datetime(prices_df['Date'].unique())
    all_dates = pd.DatetimeIndex(sorted(all_dates))
    
    # バックテスト期間
    start_date = pd.Timestamp('2016-10-01')
    end_date = pd.Timestamp('2025-10-01')
    
    trading_dates = all_dates[(all_dates >= start_date) & (all_dates <= end_date)]
    
    logger.info(f"バックテスト期間: {start_date.date()} 〜 {end_date.date()}")
    logger.info(f"営業日数: {len(trading_dates)}日")
    logger.info(f"リバランス頻度: {rebalance_frequency}営業日ごと")
    
    # 財務データのキャッシュ
    financial_cache = enhanced_financial_data.sort_values('DisclosedDate').copy()
    
    # TOPIX辞書化
    topix_dict = {}
    if not topix_df.empty:
        for _, row in topix_df.iterrows():
            topix_dict[row['Date']] = {
                'topix': row['TOPIX'],
                'sma200': row['TOPIX_SMA200']
            }
    
    # ポートフォリオ初期化
    portfolio = Portfolio(initial_capital)
    
    last_rebalance_date = None
    rebalance_counter = 0
    
    strategy_results = []
    
    # 日次ループ
    for date_idx, current_date in enumerate(tqdm(trading_dates, desc="日次バックテスト（完全版）")):
        
        # 市況フィルター
        target_positions = 20  # デフォルト
        
        if use_market_filter and current_date in topix_dict:
            topix_info = topix_dict[current_date]
            
            if topix_info['topix'] < topix_info['sma200']:
                # 市況悪化 → ポジション削減
                target_positions = 10
                logger.debug(f"{current_date.date()}: 市況フィルター発動（TOPIX < SMA200）")
        
        # リバランス判定
        should_rebalance = False
        
        if last_rebalance_date is None:
            should_rebalance = True
        elif rebalance_counter >= rebalance_frequency:
            should_rebalance = True
        
        # リバランス実行
        if should_rebalance:
            current_data = financial_cache[financial_cache['DisclosedDate'] <= current_date]
            
            if len(current_data) > 0:
                current_data = current_data.groupby('Code', as_index=False).last()
            
            if len(current_data) >= 100:
                # PBRとROEでランキング
                current_data['PBR_Rank'] = current_data['PBR'].rank(method='first', ascending=True)
                current_data['ROE_Rank'] = current_data['ROE'].rank(method='first', ascending=False)
                
                # 四分位分割
                current_data['PBR_Quartile'] = pd.qcut(current_data['PBR_Rank'], q=4, labels=[1, 2, 3, 4], duplicates='drop')
                current_data['ROE_Quartile'] = pd.qcut(current_data['ROE_Rank'], q=4, labels=[1, 2, 3, 4], duplicates='drop')
                
                # ロング候補
                long_candidates = current_data[
                    (current_data['PBR_Quartile'] == 1) & 
                    (current_data['ROE_Quartile'] == 4)
                ].nsmallest(50, 'PBR')
                
                # 現在の総資産で再計算
                current_capital = portfolio.get_total_value(price_dict, current_date)
                
                # ターゲットポートフォリオ構築
                target_portfolio = build_unit_share_portfolio(
                    long_candidates,
                    target_positions=target_positions,
                    initial_capital=current_capital
                )
                
                # リバランス実行（差分取引）
                portfolio.rebalance_to_target(
                    target_portfolio, 
                    price_dict, 
                    current_date, 
                    slippage
                )
                
                last_rebalance_date = current_date
                rebalance_counter = 0
        
        # 日次パフォーマンス記録
        total_value = portfolio.get_total_value(price_dict, current_date)
        
        daily_return = 0.0
        if date_idx > 0 and len(strategy_results) > 0:
            prev_value = strategy_results[-1]['total_value']
            if prev_value > 0:
                daily_return = (total_value - prev_value) / prev_value
        
        # 結果記録
        strategy_results.append({
            'date': current_date,
            'total_value': total_value,
            'cash': portfolio.cash,
            'daily_return': daily_return,
            'num_positions': len(portfolio.positions),
            'cumulative_tax': portfolio.total_tax_paid,
            'cumulative_trading_cost': portfolio.total_trading_cost,
            'is_rebalance_day': should_rebalance,
            'target_positions': target_positions
        })
        
        rebalance_counter += 1
    
    results_df = pd.DataFrame(strategy_results)
    transactions_df = pd.DataFrame(portfolio.transaction_history)
    
    return results_df, portfolio, transactions_df

# ===================================
# I. パフォーマンス分析
# ===================================

def calculate_annual_performance(returns: pd.Series, dates: pd.Series = None) -> float:
    """年率リターン計算"""
    if len(returns) == 0:
        return 0.0
    
    cumulative = (1 + returns).prod()
    
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
    else:
        years = len(returns) / 252  # 営業日ベース
    
    if years > 0:
        return ((cumulative ** (1/years)) - 1) * 100
    else:
        return 0.0

def calculate_volatility_annualized(returns: pd.Series, dates: pd.Series = None) -> float:
    """年率換算ボラティリティ計算"""
    if len(returns) == 0:
        return 0.0
    
    period_std = returns.std()
    
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
        periods_per_year = len(returns) / years if years > 0 else 252
    else:
        periods_per_year = 252  # 営業日ベース
    
    if periods_per_year > 0:
        return period_std * np.sqrt(periods_per_year) * 100
    else:
        return 0.0

def calculate_sharpe_ratio(returns: pd.Series, risk_free_rate: float = 0.0, dates: pd.Series = None) -> float:
    """シャープレシオ計算（年率換算）"""
    if len(returns) == 0 or returns.std() == 0:
        return 0.0
    
    excess_returns = returns - risk_free_rate / 252  # 日次無リスク金利
    mean_return = excess_returns.mean()
    std_return = excess_returns.std()
    
    if dates is not None and len(dates) >= 2:
        start_date = dates.iloc[0] if isinstance(dates.iloc[0], pd.Timestamp) else pd.Timestamp(dates.iloc[0])
        end_date = dates.iloc[-1] if isinstance(dates.iloc[-1], pd.Timestamp) else pd.Timestamp(dates.iloc[-1])
        years = (end_date - start_date).days / 365.25
        periods_per_year = len(returns) / years if years > 0 else 252
    else:
        periods_per_year = 252
    
    if std_return > 0:
        return (mean_return / std_return) * np.sqrt(periods_per_year)
    else:
        return 0.0

def calculate_max_drawdown(returns: pd.Series, dates: pd.Series = None) -> dict:
    """最大ドローダウン計算"""
    if len(returns) == 0:
        return {
            'max_drawdown': 0.0,
            'mdd_date': None,
            'mdd_period': None
        }
    
    cumulative = (1 + returns).cumprod()
    running_max = cumulative.cummax()
    drawdown = (cumulative - running_max) / running_max
    
    min_dd_idx = drawdown.idxmin()
    max_dd_value = drawdown.min() * 100
    
    mdd_date = None
    if dates is not None and len(dates) > 0:
        if min_dd_idx in dates.index:
            mdd_date = dates.loc[min_dd_idx]
        elif isinstance(min_dd_idx, (int, np.integer)) and 0 <= min_dd_idx < len(dates):
            mdd_date = dates.iloc[min_dd_idx]
    
    return {
        'max_drawdown': max_dd_value,
        'mdd_date': mdd_date,
        'mdd_period': None
    }

def calculate_additional_metrics_complete(results_df: pd.DataFrame, portfolio: Portfolio) -> dict:
    """追加パフォーマンス指標の計算（完全版）"""
    
    returns = results_df['daily_return']
    dates = results_df.get('date', None)
    
    # MDD計算
    mdd_result = calculate_max_drawdown(returns, dates)
    
    initial_capital = results_df['total_value'].iloc[0]
    final_capital = results_df['total_value'].iloc[-1]
    
    metrics = {
        'initial_capital': initial_capital,
        'final_capital': final_capital,
        'total_return': (final_capital / initial_capital - 1) * 100,
        'annual_return': calculate_annual_performance(returns, dates),
        'sharpe_ratio': calculate_sharpe_ratio(returns, risk_free_rate=0.0, dates=dates),
        'max_drawdown': mdd_result['max_drawdown'],
        'mdd_date': mdd_result['mdd_date'],
        'win_rate': (returns > 0).sum() / len(returns) * 100,
        'avg_win': returns[returns > 0].mean() * 100 if (returns > 0).any() else 0,
        'avg_loss': returns[returns < 0].mean() * 100 if (returns < 0).any() else 0,
        'volatility': calculate_volatility_annualized(returns, dates),
        'total_tax_paid': portfolio.total_tax_paid,
        'total_trading_cost': portfolio.total_trading_cost,
        'tax_rate_on_return': (portfolio.total_tax_paid / (final_capital - initial_capital)) * 100 if final_capital > initial_capital else 0,
        'num_transactions': len(portfolio.transaction_history),
    }
    
    return metrics

# ===================================
# J. 結果出力
# ===================================

def print_performance_report_complete(results_df: pd.DataFrame, metrics: dict, portfolio: Portfolio):
    """パフォーマンスレポート出力（完全版）"""
    
    print("\n" + "=" * 80)
    print("📊 日次リバランス戦略パフォーマンス結果（完全版）")
    print("=" * 80)
    
    print(f"\n【戦略設定】")
    print(f"  ポジション:          ロングのみ（ショートなし）")
    print(f"  リバランス頻度:      日次")
    print(f"  制約:                100株単位購入制限")
    print(f"  スリッページ:        0.3%")
    print(f"  税率:                20.315%")
    print(f"  市況フィルター:      有効（TOPIX < SMA200 で減少）")
    print(f"  平均ポジション数:    {results_df['num_positions'].mean():.1f}銘柄")
    
    print(f"\n【パフォーマンス】")
    print(f"  初期資本:            ¥{metrics['initial_capital']:,.0f}")
    print(f"  最終資本:            ¥{metrics['final_capital']:,.0f}")
    print(f"  累積リターン:        {metrics['total_return']:.2f}%")
    print(f"  年率リターン:        {metrics['annual_return']:.2f}%")
    print(f"  シャープレシオ:      {metrics['sharpe_ratio']:.2f}")
    print(f"  最大DD:              {metrics['max_drawdown']:.2f}%")
    if metrics.get('mdd_date'):
        mdd_date_str = metrics['mdd_date'].strftime('%Y年%m月%d日')
        print(f"  MDD発生日:          {mdd_date_str}")
    print(f"  年率ボラティリティ:  {metrics['volatility']:.2f}%")
    
    print(f"\n【コスト分析】")
    print(f"  総取引回数:          {metrics['num_transactions']:,}回")
    print(f"  累積税金:            ¥{metrics['total_tax_paid']:,.0f}")
    print(f"  累積取引コスト:      ¥{metrics['total_trading_cost']:,.0f}")
    print(f"  税金率（対利益）:    {metrics['tax_rate_on_return']:.2f}%")
    print(f"  総コスト:            ¥{metrics['total_tax_paid'] + metrics['total_trading_cost']:,.0f}")
    
    print(f"\n【勝率・損益（日次）】")
    print(f"  勝率:                {metrics['win_rate']:.1f}%")
    print(f"  平均利益:            {metrics['avg_win']:.3f}%")
    print(f"  平均損失:            {metrics['avg_loss']:.3f}%")
    
    print(f"\n【比較】")
    print(f"  年次版年率リターン:  28.52%")
    print(f"  日次版年率リターン:  {metrics['annual_return']:.2f}%")
    print(f"  差分:                {metrics['annual_return'] - 28.52:+.2f}%")
    
    print("\n" + "=" * 80)

# ===================================
# K. メイン実行
# ===================================

if __name__ == "__main__":
    
    print("=" * 80)
    print("🚀 SMBC「割安高質」戦略 - 日次リバランス版（完全版）")
    print("=" * 80)
    print(f"\n📋 検証概要:")
    print(f"  戦略名: 割安高質（ロングオンリー）")
    print(f"  検証期間: 2016年10月〜2025年10月")
    print(f"  リバランス: 日次")
    print(f"  スリッページ: 0.3%")
    print(f"  税率: 20.315%")
    print(f"  市況フィルター: 有効（TOPIX基準）")
    print(f"  実行日時: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    print("=" * 80)
    
    # データ読み込み
    statements_df = load_financial_data()
    if statements_df.empty:
        logger.error("財務データ読み込み失敗")
        exit(1)
    
    prices_df = load_existing_price_data()
    if prices_df.empty:
        logger.error("株価データ読み込み失敗")
        exit(1)
    
    # TOPIX データ読み込み
    topix_df = load_topix_data()
    
    # 財務指標計算
    enhanced_financial_data = calculate_market_metrics_fast_chunked(statements_df, prices_df, chunk_size=200)
    if enhanced_financial_data.empty:
        logger.error("財務指標計算失敗")
        exit(1)
    
    # 日次バックテスト実行（完全版）
    results, portfolio, transactions = smbc_value_quality_backtest_daily_complete(
        enhanced_financial_data, 
        prices_df,
        topix_df,
        initial_capital=10_000_000,
        rebalance_frequency=1,
        slippage=0.003,
        tax_rate=0.20315,
        use_market_filter=True
    )
    
    if results.empty:
        logger.error("バックテスト実行失敗")
        exit(1)
    
    # パフォーマンス指標計算
    metrics = calculate_additional_metrics_complete(results, portfolio)
    
    # 結果出力
    print_performance_report_complete(results, metrics, portfolio)
    
    # 結果保存
    results.to_csv('backtest_results_daily_complete.csv', index=False, encoding='utf-8-sig')
    transactions.to_csv('transactions_daily_complete.csv', index=False, encoding='utf-8-sig')
    
    logger.info("結果をCSVに保存: backtest_results_daily_complete.csv")
    logger.info("取引履歴を保存: transactions_daily_complete.csv")
    
    print("\n" + "=" * 80)
    print("🏆 日次リバランス戦略検証完了（完全版）")
    print("=" * 80)
    print(f"\n📁 出力ファイル:")
    print(f"  - バックテスト結果: backtest_results_daily_complete.csv")
    print(f"  - 取引履歴: transactions_daily_complete.csv")
    print(f"  - ログファイル: backtest_daily_value_quality_complete.log")
    print("\n")
```
